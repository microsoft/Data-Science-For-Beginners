```python
import requests
```

```python
from bs4 import BeautifulSoup
```

```python
url = 'https://en.wikipedia.org/wiki/Data_science'
```

```python
def scrape_wiki_text(url):
    # 1. Send a request to the URL
    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
    response = requests.get(url, headers = headers)

    if response.status_code != 200:
        return "Failed to retrieve the webpage."

    # 2. Parse the HTML content
    soup = BeautifulSoup(response.text, 'html.parser')

    # 3. Find the main content body
    content_div = soup.find(id = "mw-content-text")

    # 4. Extract text from all paragraph tags inside the content_div
    paragraphs = content_div.find_all('p')

    # 5. Join the paragraphs together
    article_text = " ".join([p.get_text() for p in paragraphs])

    return article_text
```

```python
text = scrape_wiki_text(url)[:500]
```

```python
wc_object = WordCloud(background_color = 'white', width = 800, height = 600, max_words = 50)
```

```python
result = wc_object.generate(text)
```
