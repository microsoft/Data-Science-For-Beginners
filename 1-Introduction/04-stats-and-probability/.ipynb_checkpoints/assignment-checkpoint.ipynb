### Task 5: Test the hypothesis that the degree of diabetes progression is different between men and women

```python
import numpy as np
import scipy.stats
```

```python
def mean_confidence_interval(data, confidence=0.95):
    a = 1.0 * data
    n = len(a)
    m, se = np.mean(a), scipy.stats.sem(a)
    h = se * scipy.stats.t.ppf((1+confidence)/2, n-1)
    return m, h
```

```python
for p in [0.85, 0.9, 0.95]:
    m1, h1 = mean_confidence_interval(df[df['SEX'] == 1]['Y'], p)
    m2, h2 = mean_confidence_interval(df[df['SEX'] == 2]['Y'], p)
    print(f"Male progression: {m1-h1:.2f}..{m1+h1:.2f}, Female progression: {m2-h2:.2f}..{m2+h2:.2f}")
```

> Confidence intervals are overlaping

```python
from scipy.stats import ttest_ind
```

```python
tval, pval = ttest_ind(df[df['SEX'] == 1]['Y'], df[df['SEX'] == 2]['Y'], equal_var=False)
print(f"t-value: {tval:.2f}, p-value: {pval:.2f}")
```

> Since p-value is above the threshold (0.05), we failed to reject Null Hypothesis (H0)
