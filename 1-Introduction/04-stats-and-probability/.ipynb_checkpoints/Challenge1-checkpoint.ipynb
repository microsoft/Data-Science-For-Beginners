# To test the hypothesis that the First_Baseman are older than Second_Baseman

## 1. Imort the dataset

```python
import pandas as pd
```

```python
df = pd.read_csv('../../data/SOCR_MLB.tsv', sep = '\t')
df
```

### Now, we'll rename the columns.

```python
df.columns = ['Name', 'Team', 'Role', 'Height', 'Weight', 'Age']
df
```

## 2. Calculate mean confidence interval

```python
import scipy.stats
```

```python
import numpy as np
```

```python
def mean_confidence_interval(data, confidence=0.95):
    a = 1.0 * data
    n = len(a)
    m, se = np.mean(a), scipy.stats.sem(a)
    h = scipy.stats.t.ppf((1+confidence)/2, n-1) * se
    return m, h
```

```python
df.groupby('Role').agg(
    Mean_Weight = ('Weight', 'mean'),
    Mean_Height = ('Height', 'mean'),
    Mean_Age = ('Age', 'mean')
)
```

```python
for p in [0.85, 0.9, 0.95]:
    m1, h1 = mean_confidence_interval(df.loc[df['Role'] == 'First_Baseman', ['Age']], p)
    m2, h2 = mean_confidence_interval(df.loc[df['Role'] == 'Second_Baseman', ['Age']], p)
    print(f"p: {p:.2f}, 1st_Baseman_Height: {m1 - h1[0]:.2f}..{m1 + h1[0]:.2f}, 2nd_Baseman_Height: {m2 - h2[0]:.2f}..{m2 + h2[0]:.2f}")
```

## 3. Student's t-test

```python
from scipy.stats import ttest_ind
```

```python
tval, pval = ttest_ind(df.loc[df['Role'] == 'First_Baseman', ['Age']], df.loc[df['Role'] == 'Second_Baseman', ['Age']], equal_var=False)
print(f"t-val: {tval[0]:.2f}, p-val: {pval[0]:.2f}")
```

> Since, p-val = 0.60 (greater than 0.05). It fails to reject the Null Hypothesis (H0)
