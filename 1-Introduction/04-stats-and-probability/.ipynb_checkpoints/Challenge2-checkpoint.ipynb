```python
import pandas as pd
```

```python
df = pd.read_table('../../data/SOCR_MLB.tsv')
df
```

```python
df.columns = ['Name', 'Team', 'Role', 'Weight', 'Height', 'Age']
df
```

```python
import numpy as np
import scipy.stats
```

```python
def mean_confidence_interval(data, confidence=0.95):
    a = 1.0 * data
    n = len(a)
    m, se = np.mean(a), scipy.stats.sem(a)
    h = se * scipy.stats.t.ppf((1+confidence)/2, n-1)
    return m, h
```

```python
df.groupby('Role').agg(
    Mean_Height=('Height', 'mean'),
    Count=('Age', 'count')
)
```

```python
for p in [0.85, 0.9, 0.95]:
    m1, h1 = mean_confidence_interval(df.loc[df['Role'] == 'First_Baseman', ['Height']], p)
    m2, h2 = mean_confidence_interval(df.loc[df['Role'] == 'Third_Baseman', ['Height']], p)
    print(f"p: {p}, 1st_Baseman_Height: {m1-h1[0]:.2f}..{m1+h1[0]:.2f}, 3rd_Baseman_Height: {m2-h2[0]:.2f}..{m2+h2[0]:.2f}")
```

# To test the hypothesis that first basemen are taller than third basemen

## We'll do the Student's t-test

```python
from scipy.stats import ttest_ind
```

```python
tval, pval = ttest_ind(df.loc[df['Role'] == 'First_Baseman', 'Height'], df.loc[df['Role'] == 'Third_Baseman', 'Height'], equal_var=False)
print(f"t_val: {tval:.2f}, p-value: {pval:.2f}")
```

> Since p-value is zero, we failed to reject the Alternate Hypothesis (H1)
