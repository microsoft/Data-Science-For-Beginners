## ယုံကြည်မှု အကွာအဝေးများ

ယခုအခါ ဂိမ်းကစားသမားများ၏ အလေးချိန်နှင့် အရှည်တို့အတွက် ယုံကြည်မှု အကွာအဝေးများကို တွက်ချက်ကြမည်။ ကျနော်တို့သည် ကုဒ်ကို [ဤ stackoverflow ဆွေးနွေးမှုမှ](https://stackoverflow.com/questions/15033511/compute-a-confidence-interval-from-sample-data) အသုံးပြုမည်ဖြစ်သည်။


```python
import scipy.stats

def mean_confidence_interval(data, confidence=0.95):
    a = 1.0 * np.array(data)
    n = len(a)
    m, se = np.mean(a), scipy.stats.sem(a)
    h = se * scipy.stats.t.ppf((1 + confidence) / 2., n-1)
    return m, h

for p in [0.85, 0.9, 0.95]:
    m, h = mean_confidence_interval(df['Weight'].ffill(),p)
    print(f"p={p:.2f}, mean = {m:.2f} ± {h:.2f}")
```

## သံသယစမ်းသပ်ခြင်း

ကျွန်ုပ်တို့၏ဘေ့စ်ဘောကစားသမားဒေတာစုစည်းမှုအတွင်း မတူကွဲပြားသော အခန်းကဏ္ဍများကို ရှာဖွေကြည့်ရအောင်။


```python
df.groupby('Role').agg({ 'Weight' : 'mean', 'Height' : 'mean', 'Age' : 'count'}).rename(columns={ 'Age' : 'Count'})
```

ပထမမြောက် ဘေ့စ်မင်များသည် ဒုတိယမြောက် ဘေ့စ်မင်များထက် များမြင့်ကြောင်း သီအိုရီကို စမ်းသပ်ကြည့်လိုပါစို့။ ဒါကို လုပ်ဖို့အတွက် ရိုးရှင်းဆုံး နည်းလမ်းမှာ ယုံကြည်မှု အတွင်းခွင့်များကို စမ်းသပ်ခြင်းဖြစ်ပါတယ်။


```python
for p in [0.85,0.9,0.95]:
    m1, h1 = mean_confidence_interval(df.loc[df['Role']=='First_Baseman',['Height']],p)
    m2, h2 = mean_confidence_interval(df.loc[df['Role']=='Second_Baseman',['Height']],p)
    print(f'Conf={p:.2f}, 1st basemen height: {m1-h1[0]:.2f}..{m1+h1[0]:.2f}, 2nd basemen height: {m2-h2[0]:.2f}..{m2+h2[0]:.2f}')
```

ကျွန်ုပ်တို့သည် အချိန်ကာလများ မတူညီကြောင်း တွေ့နိုင်ပါသည်။

အယူအဆကို အတည်ပြုရန် စာရင်းအင်းပိုမို မှန်ကန်သောနည်းလမ်းမှာ **Student t-test** ကို အသုံးပြုခြင်း ဖြစ်ပါသည်။


```python
from scipy.stats import ttest_ind

tval, pval = ttest_ind(df.loc[df['Role']=='First_Baseman',['Height']], df.loc[df['Role']=='Second_Baseman',['Height']],equal_var=False)
print(f"T-value = {tval[0]:.2f}\nP-value: {pval[0]}")
```

`ttest_ind` function က ပြန်လည်ပေးသော တန်ဖိုးနှစ်ခုမှာ -
* p-value ကို အလားအလာအဖြစ် ကျော်ကြားသော မိန်းနှစ်ခုဟာ ပြန်လည်တက်သော တူညီသော အလိုတူတူရှိမရှိနှင့် ဆက်နွယ်သည်ဟု မှတ်သတ်နိုင်သည်။ ကျွန်ုပ်တို့ကိစ္စတွင် အလွန်နည်းပါးသောအရာဖြစ်ပြီး၊ ပထမဆုံး basemen တွေ ပိုမြင့်သည်ဆိုသော အထောက်အထား အလွန်ခိုင်မာကြောင်းကို ပြောသည်။
* t-value သည် t-test တွင် သုံးသော normalized mean difference ၏ အလယ်တန်းတန်ဖိုးဖြစ်ပြီး၊ သတ်မှတ်ထားသော ယုံကြည်မှုတန်ဖိုးအတွက် threshold တန်ဖိုးနှင့် နှိုင်းယှဉ်သည်။


## အလယ်ဗဟို ကန့်သတ်ချက် သီအိုရီဖြင့် ပုံမှန်ဖြန့်ဖြူးမှုကို အတုယူခြင်း

Python အတွင်းရှိ အတုအစွန်းထုတ်ယူပေးသူသည် uniform ဖြန့်ဖြူးမှုပေးရန်ဒီဇိုင်းရေးဆွဲထားသည်။ ပုံမှန်ဖြန့်ဖြူးမှုအတွက် generator တစ်ခု ဖန်တီးလိုပါက အလယ်ဗဟို ကန့်သတ်ချက် သီအိုရီကို အသုံးပြုနိုင်သည်။ ပုံမှန်ဖြန့်ဖြူးသော တန်ဖိုးတစ်ခုရရန် အသုံးပြုသူသည် uniform ဖြန့်ဖြူးသည့် စမ်းသပ်ဖြင့် ပျမ်းမျှတန်ဖိုးကို တွက်ချက်ပေးရန်သာ လိုအပ်မည်။


```python
def normal_random(sample_size=100):
    sample = [random.uniform(0,1) for _ in range(sample_size) ]
    return sum(sample)/sample_size

sample = [normal_random() for _ in range(100)]
plt.figure(figsize=(10,6))
plt.hist(sample)
plt.tight_layout()
plt.show()
```

## ဆက်စပ်မှုနှင့် လူပုန် Baseball Corp

ဆက်စပ်မှုက ကျွန်ုပ်တို့အား ဒေတာ စီးရီးများအကြား ဆက်နွယ်မှုကို ရှာဖွေရန် ခွင့်ပြုသည်။ ကျွန်ုပ်တို့၏ ကလေးကစားနမူနာတွင် လူပုန် Baseball ကော်ပိုရေးရှင်းတစ်ခုရှိပြီး ကစားသမားများ၏အမြင့်အတိုင်း ငွေကြေးပေးဆောင်သောကြေညာကြပါစို့ - ကစားသမားအမြင့်ပိုများလျှင် ပိုမိုငွေ ရရှိမည်။ အခြေခံလစာ $1000 ရှိပြီး၊ အမြင့်အပေါ် မှီ၍ $0 မှ $100 အထိ ပိုလစာ ရရှိနိုင်ပါသည်။ MLB မှ ကိုယ်တိုင် ကစားသမားများကို ယုံကြည်ပါသည်၊ ၎င်းတို့၏ မစွဲလမ်းသော လစာကိုတွက်ချက်ကြမည်။


```python
heights = df['Height'].ffill()
salaries = 1000+(heights-heights.min())/(heights.max()-heights.mean())*100
print(list(zip(heights, salaries))[:10])
```

ယခုအခါ ၎င်းသက်ဆိုင်ရာလိုက်လျောမှုများ၏ covariance နဲ့ correlation ကိုတွက်ဆကြမယ်။ `np.cov` က ကျွန်တော်တို့ကို **covariance matrix** လို့ခေါ်တဲ့ matrix တစ်ခုကို ပေးပါလိမ့်မယ်၊ ဒါဟာ covariance ကို မတူညီတဲ့ပြောင်းလဲနိုင်မှုအနေနဲ့ တိုးချဲ့ထားတာပါ။ covariance matrix $M$ ရဲ့ အခြေခံအစိတ် $M_{ij}$ က input ပြောင်းလဲနိုင်မှု $X_i$ နဲ့ $X_j$ အကြား သက်ဆိုင်မှု ဖြစ်ပါတယ်၊ ဒါ့အပြင် diagonal မှာရှိတဲ့ $M_{ii}$ တွေက $X_{i}$ ရဲ့ variance ဖြစ်ပါတယ်။ အလားတူပဲ `np.corrcoef` က ကျွန်တော်တို့ကို **correlation matrix** ကို ပေးပါလိမ့်မယ်။


```python
print(f"Covariance matrix:\n{np.cov(heights, salaries)}")
print(f"Covariance = {np.cov(heights, salaries)[0,1]}")
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

ကိုယ်စားပြုချက်တစ်ခု ၁ နှင့်တူညီသည်မှာ ဤပြသာနာနှစ်ခုအကြား **ဇယားလိုက် ဆက်နွယ်မှု** ပြင်းထန်ကြောင်းကို ဖော်ပြသည်။ တန်ဖိုးတစ်ခုကို အခြားတန်ဖိုးနှင့် ထိုးမြှောက်ခြင်းအားဖြင့် ဇယားလိုက် ဆက်နွယ်မှုကို မြင်တွေ့နိုင်သည်။


```python
plt.figure(figsize=(10,6))
plt.scatter(heights,salaries)
plt.tight_layout()
plt.show()
```

ဆက်နွယ်မှုသည် လိုင်းဟာရ မဖြစ်ပါက ဖြစ်ပေါ်လာနိုင်သည့်အခြေအနေကို ကြည့်ကြရအောင်။ ကျွန်ုပ်တို့၏ ကော်ပိုရေးရှင်းသည် အမြင့်နှင့် လစာများ အကြား အလွယ်တကူ တွက်နိုင်သော လိုင်းဟာရ လှုပ်ရှားမှုကို ဖုံးကွယ်ရန်ဆုံးဖြတ်ခဲ့ပြီး၊ `sin` ကဲ့သို့သော မလိုင်းဟာရ အစိတ်အပိုင်းတစ်ခုကို ဖော်ထုတ်ခဲ့သည်ဟု သတ်မှတ်ကြပါစို့။


```python
salaries = 1000+np.sin((heights-heights.min())/(heights.max()-heights.mean()))*100
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

ဤအဖြစ်တွင် ဆက်စပ်မှုသည် နည်းနည်းနည်းသောဖြစ်ပေမယ့် များမြင့်ဆဲဖြစ်သည်။ ယခုတွင် ဆက်နွယ်မှုကို နည်းနည်း ပို၍ ပွားသိပ်ရှင်းမှုမရှိအောင် ဝင်ငွေသို့ ကျပ်တည်းမှုအပိုဖြစ်စေသော အလိုမရှိသော သို့မဟုတ် ရွေးချယ်ခြင်း မရှိသည့် အပြောင်းအလဲအချိုးအစားတစ်ခု ထည့်သွင်းလိုပါက ဖြစ်လာမည့်အခြေအနေကို ကြည့်ကြရအောင်-  


```python
salaries = 1000+np.sin((heights-heights.min())/(heights.max()-heights.mean()))*100+np.random.random(size=len(heights))*20-10
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

```python
plt.figure(figsize=(10,6))
plt.scatter(heights, salaries)
plt.tight_layout()
plt.show()
```

> ဒီအတိုင်း ဒါ့နောက် အဝိုင်းလေးတွေဟာ ထောင့်တန်းလိုက် စုစည်းထားတာ ဘာကြောင့်လဲ ထင်နိုင်ပါသလား?

ကျွန်ုပ်တို့က လစာလို အတုအယောင်ထုတ် ထုတ်လုပ်ထားတဲ့ အယူအဆတစ်ခုနဲ့ မှတ်သားထားတဲ့ အမျိုးအစား *ထောင့်* အကြား ဆက်စပ်မှုကို တွေ့ရှိနိုင်ခဲ့ပါတယ်။ ထို့အပြင် ထောင့်နဲ့ အလေးချိန်ကဲ့သို့သော မှတ်သားထားတဲ့ အမျိုးအစား နှစ်ခုမှာလည်း ဆက်စပ်မှု ရှိမရှိ ကြည့်လိုက်ကြရအောင်။


```python
np.corrcoef(df['Height'].ffill(),df['Weight'])
```

မန္တလေး၊ ကျွန်ုပ်တို့သည် ရလဒ်အနည်းငယ်မရရှိခဲ့ပါ - `nan` တစ်ချို့သာ ရှိသည်။ ၎င်းသည် ကျွန်ုပ်တို့၏ စီးရီးအချို့တွင် အတိအကျမသတ်မှတ်ထားသော တန်ဖိုးများရှိပြီး၊ ၎င်းကို `nan` ဖြင့်ဖော်ပြထားသည်၊ ၎င်းသည် လုပ်ဆောင်ချက်ရလဒ်ကိုလည်း အတိအကျမသတ်မှတ်နိုင်စေသည်။ မက်ထရစ်စ်အကြောင်းကြည့်ရှုသည်မှာ `Weight` သည် ပြဿနာရှိသော ကော်လံဖြစ်သည်၊ ကာလအတွင်း `Height` တန်ဖိုးများအကြား ကိုယ်တွဲဆက်နွယ်မှုကိုတွက်ချက်ပြီးဖြစ်သည်။

> ဤဥပမာသည် **ဒေတာပြင်ဆင်ခြင်း** နှင့် **သန့်ရှင်းခြင်း** ၏ အရေးပါမှုကို ပြသသည်။ သင့်တော်သော ဒေတာမရှိပါက အရာဝတ္ထုကိုမတွက်ချက်နိုင်ပါ။

ကွက်လပ်တန်ဖိုးများကို ဖြည့်ရန် `fillna` နည်းလမ်းကို အသုံးပြု၍ ကိုယ်တွဲဆက်နွယ်မှုကိုတွက်ချက်ကြပါစို့ - 


```python
np.corrcoef(df['Height'].ffill(), df['Weight'])
```

တကယ်တော့ ဆက်နွယ်မှုတစ်ခုရှိပါတယ်၊ ဒါပေမယ့် ကျွန်တော်တို့ရဲ့ စက်မှုလုပ်ငန်းနမူနာအတိုင်း မဟုတ်ပါတယ်။ တကယ်လို့ တစ်ခုစီတန်ဖိုးနှစ်ခုအကြား စကက်တာပလာကို ကြည့်မယ်ဆိုရင် ဆက်နွယ်မှုဟာ ပိုပြီးရှင်းလင်းမှုနည်းပါတယ်။ 


```python
plt.figure(figsize=(10,6))
plt.scatter(df['Weight'],df['Height'])
plt.xlabel('Weight')
plt.ylabel('Height')
plt.tight_layout()
plt.show()
```
