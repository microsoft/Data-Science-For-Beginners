# အခက်အခဲ: ဒေတာသိပ္ပံအကြောင်း မက်စ်တစ်ခု စိစစ်ခြင်း

ဤဥပမာတွင်၊ ရိုးရာ ဒေတာသိပ္ပံ လုပ်ငန်းစဉ်၏ အဆင့်များအားလုံးဖြင့် ရိုးရှင်းသောလေ့ကျင့်ခန်းတစ်ခု မျှဝေပါမည်။ သင်သည် ကုဒ်ဘယ်လိုရေးရည်မထားဘဲ အောက်တွင်ရှိသောဆဲလ်များကို နှိပ်ပြီး လည်ပတ်မှုကို စမ်းကြည့်နိုင်ပြီး နောက်ထွက်ကို ကြည့်ရှုနိုင်သည်။ အခက်အခဲအနေဖြင့်၊ သင်သည် ထိုကုဒ်ကို ကွဲပြားသောဒေတာနှင့် စမ်းသပ်ကြည့်ရန်အားပေးပါသည်။

## ရည်မှန်းချက်

ဤသင်ခန်းစာတွင်၊ ဒေတာသိပ္ပံနှင့် ပတ်သက်သော အကြောင်းအရာများကို ဆွေးနွေးခဲ့ပြီးဖြစ်သည်။ **စာသားဖြာခြင်း** ကိုပြုလုပ်ကာ ပိုမိုဆက်စပ်နေသော အကြောင်းအရာများကို ရှာဖွေရန်ကြိုးစားကြစို့။ ဒေတာသိပ္ပံအကြောင်း စာသားတစ်ခုကို စတင်ယူပြီး၊ အဓိကစကားလုံးများထုတ်ယူကာ နောက်မှ ရလဒ်ကို မှတ်ဉာဏ်ဖြင့် ဖော်ပြကြမည်။

စာသားအဖြစ် Wikipedia တွင်ရှိသည့် ဒေတာသိပ္ပံ စာမျက်နှာကို အသုံးပြုမည်။


```python
url = 'https://en.wikipedia.org/wiki/Data_science'
```

## အဆင့် ၁: ဒေတာရယူခြင်း

ဒေတာအသိပညာလုပ်ငန်းစဉ်တိုင်းအတွက် ပထမဆုံးအဆင့်မှာ ဒေတာကို ရယူခြင်းဖြစ်သည်။ အဲဒီအတွက် `requests` စာကြည့်တိုက်ကို အသုံးပြုမည်ဖြစ်သည်။


```python
import requests

# Define a custom header.
headers = {
    'User-Agent': 'DataScienceChallenge/1.0 (myemail@gmail.com)'
}

# Pass the headers into the get request
response = requests.get(url, headers=headers)

if response.status_code == 200:
    text = response.content.decode('utf-8')
    print(text[:1000])
else:
    print(f"Error: {response.status_code}")
```

## အဆင့် ၂: ဒေတာကို ပြောင်းလဲခြင်း

နောက်တစ်ဆင့်မှာ မှတ်ချက်ဖြေရှင်းရန်သင့်လျော်သော ပုံစံသို့ ဒေတာကို ပြောင်းလဲခြင်း ဖြစ်သည်။ ကျွန်ုပ်တို့အနေဖြင့် ဤမူလအနေဖြင့် ဝက်ဘ်စာမျက်နှာမှ HTML ကုဒ်ကို ဒေါင်းလုတ်လုပ်ပြီး ပြင်ပစာသားသို့ ပြောင်းလဲရန် လိုအပ်သည်။

ဒီလူကြိုက်များသော Python စာကြည့်တော် [BeautifulSoup](https://www.crummy.com/software/BeautifulSoup/) ကို သုံး၍ ပြုလုပ်နိုင်သည်။ BeautifulSoup သည် HTML အစိတ်အပိုင်းများအား ဦးတည်ရန် ခွင့်ပြုသဖြင့် Wikipedia ပင်မဆောင်းပါးအကြောင်းအရာများကို အာရုံစိုက်ပြီး လမ်းညွှန်မီနူးများ၊ ဘေးဘားများ၊ ခြေတုပ်များနဲ့ အလွဲအလွန် မဆက်စပ်သော အကြောင်းအရာများကို လျှော့ချနိုင်သည် (သို့သော် အချို့သော စာသားများ ကွက်တိများသည် မကျန်ရစ်နိုင်ပါ)။


ပထမဦးစွာ HTML ဖော်မတ်ချရေးအတွက် BeautifulSoup လိုက်ဘရေရီကို တပ်မတ်ရပါမည်။


```python
import sys
!{sys.executable} -m pip install beautifulsoup4
```

```python
from bs4 import BeautifulSoup

# Parse the HTML content
soup = BeautifulSoup(text, 'html.parser')

# Extract only the main article content from Wikipedia
# Wikipedia uses 'mw-parser-output' class for the main article content
content = soup.find('div', class_='mw-parser-output')

def clean_wikipedia_content(content_node):
    """Remove common non-article elements from a Wikipedia content node."""
    # Strip jump links, navboxes, reference lists/superscripts, edit sections, TOC, sidebars, etc.
    selectors = [
        '.mw-jump-link',
        '.navbox',
        '.reflist',
        'sup.reference',
        '.mw-editsection',
        '.hatnote',
        '.metadata',
        '.infobox',
        '#toc',
        '.toc',
        '.sidebar',
    ]
    for selector in selectors:
        for el in content_node.select(selector):
            el.decompose()

if content:
    # Clean the content node to better approximate article text only.
    clean_wikipedia_content(content)
    text = content.get_text(separator=' ', strip=True)
    print(text[:1000])
else:
    print("Could not find main content. Using full page text.")
    text = soup.get_text(separator=' ', strip=True)
    print(text[:1000])
```

## အဆင့် ၃: အမြင်ရရန်

အရေးကြီးဆုံး အဆင့်မှာ ကျွန်ုပ်တို့၏ ဒေတာကို အမြင်ရရှိနိုင်သော ပုံစံတစ်ခုအဖြစ် ပြောင်းလဲခြင်းဖြစ်သည်။ ကျွန်ုပ်တို့၏ ကိစ္စမှာ စာသားမှ keyword များကို ထုတ်ယူပြီး ဘယ် keyword များက ပိုမိုအဓိပ္ပါယ်ရှိသည်ကို ကြည့်ရှုလိုသည်။

keyword extraction အတွက် Python library တစ်ခုဖြစ်သည့် [RAKE](https://github.com/aneesha/RAKE) ကို အသုံးပြုမည်။ အရင်ဆုံး သက်ဆိုင်ရာ library မရှိသေးပါက ထည့်သွင်းလိုက်ပါ။ 


```python
import sys
!{sys.executable} -m pip install nlp_rake
```

အဓိက လုပ်ဆောင်ချက်ဟာ `Rake` အရာဝတ္ထုပေါ်မှာ ရရှိနိုင်ပြီး၊ ကျွန်တော်တို့က တချို့ပါတီမီတာတွေကို အသုံးပြုပြီး စိတ်ကြိုက်ပြင်ဆင်နိုင်ပါတယ်။ ကျွန်တော်တို့အတွက်တော့ စကားလုံးအနည်းဆုံး အရှည်ကို ၅ လုံး၊ စာရွက်ထဲ စကားလုံးအားဖြင့် အနည်းဆုံး ထပ်ခါထပ်ခါ ပေါ်ရှိမှုကို ၃ ဖြစ်အောင်၊ စကားလုံးတွဲ အများဆုံး စကားလုံးခြားကို ၂ ဖြစ်အောင် သတ်မှတ်မှာဖြစ်ပါတယ်။ အခြားတန်ဖိုးတွေနဲ့လည်း စမ်းသပ်ပြီး ဖော်ပြချက်ကို ကြည့်ရှုနိုင်ပါသည်။


```python
import nlp_rake
extractor = nlp_rake.Rake(max_words=2,min_freq=3,min_chars=5)
res = extractor.apply(text)
res
```


သင်္ချာဆက်စပ်အရေးကြီးမှုအဆင့်နှင့်အတူ စကားလုံးစာရင်းကို ရရှိခဲ့သည်။ မြင်နိုင်သကဲ့သို့ စက်သင်ယူမှုနှင့် သိပ္ပံဒေတာကဲ့သို့ အရေးကြီးဆုံးဘာသာရပ်များသည် စာရင်းတွင် ထိပ်တန်းနေရာများတွင် ပါရှိနေသည်။

## အဆင့် ၄: ရလဒ်ကို မြင်ကွင်းဖော်ခြင်း

လူများသည် ဒေတာကို မြင်ကွင်းပုံစံဖြင့် အကောင်းဆုံးဖော်ထုတ်နိုင်ကြသည်။ ထို့ကြောင့် ဒေတာအား မြင်ကွင်းဖော်ရန် ရည်ရွယ်ချက်ဖြင့် ဝေါဟာရများ၏ မန်စာဖြန့်ဝေမှုကို Python တွင် `matplotlib` စာကြည့်တိုက်ကိုအသုံးပြုပြီး ပုံဆွဲနိုင်သည်။


```python
import matplotlib.pyplot as plt

def plot(pair_list):
    k,v = zip(*pair_list)
    plt.bar(range(len(k)),v)
    plt.xticks(range(len(k)),k,rotation='vertical')
    plt.show()

plot(res)
```

သို့သော်စကားလုံးများ၏ကြိမ်နှုန်းများကိုမြင်သာအောင်ပြသရာတွင် ပိုမိုကောင်းမွန်သောနည်းလမ်းတစ်ခုရှိသည် - **Word Cloud** ကိုအသုံးပြုခြင်းဖြစ်သည်။ ကျွန်ုပ်တို့၏ keyword စာရင်းမှ စကားလုံးမိုးကို ရိုက်နှိပ်ရန် အခြားစာကြည့်တိုက်တစ်ခုကို ထည့်သွင်းထည့်သွင်းရန် လိုအပ်ပါမည်။


```python
!{sys.executable} -m pip install wordcloud
```

`WordCloud` object သည် မူလစာသားကို သို့မဟုတ်၊ စကားလုံးများနှင့် ၎င်းတို့၏ ကြိမ်နှုန်းများပါသော အစီအစဉ်စာရင်းကို လက်ခံပြီး၊ ပုံတစ်ပုံပြန်လည်ထုတ်ပေးသည်၊ ၎င်းကို `matplotlib` သုံး၍ ပြသနိုင်ပါသည်။


```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

wc = WordCloud(background_color='white',width=800,height=600)
plt.figure(figsize=(15,7))
plt.imshow(wc.generate_from_frequencies({ k:v for k,v in res }))
```

ကျွန်ုပ်တို့သည် `WordCloud` သို့ မူရင်းစာသားကိုလည်း ပေးပို့နိုင်သည် - ကျွန်ုပ်တို့သည် ဆင်တူရလဒ်ကို ရနိုင်ကြောင်း ကြည့်ကြပါစို့။ 


```python
plt.figure(figsize=(15,7))
plt.imshow(wc.generate(text))
```

သင်သည် word cloud သည် ယခုပိုမိုအံ့အားသင့်ဖွယ်ရှိသည်ကို တွေ့မြင်နိုင်ပြီး၊ သို့ရာတွင် အနှစ်သာရမရှိသော စကားလုံးများ (ဥပမာ- `Retrieved on` ကဲ့သို့ ဆက်စပ်မထားသော စကားလုံးများ) များလည်း ပါဝင်နေပါသည်။ ထို့အပြင် နှစ်စကားဖြစ်သော ကီးဝို့(ပ်)များ (ဥပမာ- *data scientist*, သို့မဟုတ် *computer science*) ကိုလည်း နည်းပါးသွားပါသည်။ ၎င်းသည် RAKE အယ်လဂိုရီသမ်သည် စာသားမှ ကောင်းမွန်သော ကီးဝေါ့(ပ်)များကို ရွေးချယ်ရာတွင် ပိုမိုကောင်းမွန်သော အလုပ်လုပ်နိုင်ခြင်းကြောင့်ဖြစ်သည်။ ဤဥပမာသည် ဒေတာ မတည့်မှုနှင့် သန့်ရှင်းမှု အရေးပါမှုကို ဖော်ပြပြသပါသည်၊ ၏အဆုံးတွင် ပေါ်လွင်သောပုံရိပ်ကြောင့် ကျွန်ုပ်တို့သည် ပိုမိုကောင်းမွန်သော ဆုံးဖြတ်ချက်များ ချနိုင်မည်ဖြစ်ပါသည်။

ဤလေ့ကျင့်ခန်း၌ ကျွန်ုပ်တို့သည် Wikipedia စာသားမှ အဓိပ္ပါယ်တစ်ချို့ကို ရရှိအောင် လုပ်ဆောင်ရာတွင် ကီးဝေါ့(ပ်)များနှင့် word cloud ပုံစံဖြင့် ရှာဖွေခြင်း လုပ်ငန်းစဉ် တစ်ခု လွယ်ကူစွာ ဖြတ်သန်းခဲ့ပါသည်။ ဤဥပမာသည် လွယ်ကူသော်လည်း ဒေတာသိပ္ပံပညာရှင်တစ်ဦး သည် ဒေတာဆိုင်ရာ အလုပ်များလုပ်သောအခါ ပုံမှန်ခရီးစဉ် များအားလုံးကို ကောင်းစွာ ပြသပေးပါသည်၊ ဒေတာဆောင်ရွက်မှုမှစ၍ ရှုထောင့်ပြခြင်းအထိ ဖြစ်ပါသည်။

ကျွန်ုပ်တို့၏ သင်တန်းတွင် သင်္ကေတများအားလုံးကို အသေးစိတ်ဆွေးနွေးသွားမည် ဖြစ်ပါသည်။
