## ជួរជឿជាក់

ឥឡូវនេះ យើងនឹងគណនាជួរជឿជាក់សម្រាប់ទម្ងន់ និងកម្ពស់របស់អ្នកលេងបេស្បាល់។ យើងនឹងប្រើកូដ [ពីការពិភាក្សានៅ stackoverflow นี้](https://stackoverflow.com/questions/15033511/compute-a-confidence-interval-from-sample-data)៖


```python
import scipy.stats

def mean_confidence_interval(data, confidence=0.95):
    a = 1.0 * np.array(data)
    n = len(a)
    m, se = np.mean(a), scipy.stats.sem(a)
    h = se * scipy.stats.t.ppf((1 + confidence) / 2., n-1)
    return m, h

for p in [0.85, 0.9, 0.95]:
    m, h = mean_confidence_interval(df['Weight'].ffill(),p)
    print(f"p={p:.2f}, mean = {m:.2f} ± {h:.2f}")
```

## ការធ្វើតេស្តសន្និដ្ឋាន

ចូរយើងស្វែងយល់អំពីតួនាទីផ្សេងៗនៅក្នុងទិន្នន័យអ្នកលេងបេសបលរបស់យើង៖


```python
df.groupby('Role').agg({ 'Weight' : 'mean', 'Height' : 'mean', 'Age' : 'count'}).rename(columns={ 'Age' : 'Count'})
```

មកសាកល្បងសន្មតការណ៍ថា អ្នកលេងទីមួយនៅទីមូលដ្ឋានខូចខ្លាំងជាងអ្នកលេងទីពីរនៅទីមូលដ្ឋាន។ វិធីងាយស្រួលបំផុតក្នុងការធ្វើនេះគឺសាកល្បងចន្លោះទំនុកចិត្ត៖


```python
for p in [0.85,0.9,0.95]:
    m1, h1 = mean_confidence_interval(df.loc[df['Role']=='First_Baseman',['Height']],p)
    m2, h2 = mean_confidence_interval(df.loc[df['Role']=='Second_Baseman',['Height']],p)
    print(f'Conf={p:.2f}, 1st basemen height: {m1-h1[0]:.2f}..{m1+h1[0]:.2f}, 2nd basemen height: {m2-h2[0]:.2f}..{m2+h2[0]:.2f}')
```

យើងអាចឃើញថា ចន្លោះពេលនីមួយៗមិនទាន់ប្រហែលគ្នា។

វិធីសាស្ត្រមានស្ថិតិដែលមានភាពត្រឹមត្រូវជាងក្នុងការបញ្ជាក់សម្រង់គំនិតគឺការប្រើប្រាស់ **ការធ្វើតេស្ត​​Student t**:


```python
from scipy.stats import ttest_ind

tval, pval = ttest_ind(df.loc[df['Role']=='First_Baseman',['Height']], df.loc[df['Role']=='Second_Baseman',['Height']],equal_var=False)
print(f"T-value = {tval[0]:.2f}\nP-value: {pval[0]}")
```

តម្លៃពីរដែលបានត្រឡប់មកវិញដោយមុខងារ `ttest_ind` គឺ:
* តម្លៃ p-value អាចត្រូវបានគេពិចារណាថា ជាភាគហ៊ុននៃការហ្វឹកហាត់ពីរដែលមានមធ្យមដូចគ្នា។ ក្នុងករណីរបស់យើង វាតិចណាស់ មានន័យថាមានភស្តុតាងដ៏ខ្លាំងគាំទ្រថាអ្នកលេងបាស្មែនទីមួយចម្បងមានកម្ពស់វែងជាង។
* តម្លៃ t-value គឺជាតម្លៃមធ្យមកំណត់ទំរង់ធម្មតាចន្លោះដែលត្រូវបានប្រើនៅក្នុងការធ្វើតេស្ត t ហើយវាត្រូវបានប្រៀបធៀបជាមួយតម្លៃស្តង់ដារសម្រាប់តម្លៃទំនុកចិត្តដែលបានកំណត់។


## ការប្រុផ្សាយចែកចាយធម្មតាជាមួយទ្រឹស្ដីកម្រិតកណ្ដាល

ម៉ាស៊ីនចេតនា​សំណុំ​បែប​រ៉ែន​ដុំ​នៅក្នុង Python ត្រូវ​បាន​រចនា​ឡើង​ដើម្បីផ្តល់ឲ្យ​ពួក​យើង​នូវចែកចាយ​ឯករាជ្យស្របគ្នា។ ប្រសិន​បើយើងចង់បង្កើតម៉ាស៊ីនចេញសម្រាប់ចែកចាយធម្មតា យើងអាចប្រើទ្រឹស្តីកម្រិតកណ្ដាល។ ដើម្បីទទួលបានតម្លៃចែកចាយធម្មតា យើងគ្រាន់តែគណនាមធ្យមនៃគំរូបង្កើតដោយឯករាជ្យស្របគ្នា។ 


```python
def normal_random(sample_size=100):
    sample = [random.uniform(0,1) for _ in range(sample_size) ]
    return sum(sample)/sample_size

sample = [normal_random() for _ in range(100)]
plt.figure(figsize=(10,6))
plt.hist(sample)
plt.tight_layout()
plt.show()
```

ការតភ្ជាប់អនុញ្ញាតឱ្យយើងរកឃើញទំនាក់ទំនងរវាងខ្សែដាតា។ នៅក្នុងឧទាហរណ៍ល្បែងរបស់យើង សូមជម្រាបថាមានក្រុមហ៊ុនបេសបូលអាក្រក់មួយដែលបង់ប្រាក់ទៅកាន់អ្នកលេងរបស់ខ្លួនផ្អែកលើកម្ពស់របស់ពួកគេ - អ្នកលេងដែលខ្ពស់ជាងនឹងទទួលបានប្រាក់ច្រើនជាង។ សន្និដ្ឋានថាមានប្រាក់បៀវត្សចំបង $1000 ហើយមានប្រាក់បន្ថែមពី $0 ដល់ $100 ដែលផ្អែកលើកម្ពស់។ យើងនឹងយកអ្នកលេងពិតពី MLB ហើយគណនាប្រាក់ឈ្នួលស្រមៃរបស់ពួកគេ៖


```python
heights = df['Height'].ffill()
salaries = 1000+(heights-heights.min())/(heights.max()-heights.mean())*100
print(list(zip(heights, salaries))[:10])
```

ឥឡូវនេះ យើងនឹងគណនាអាចប្រើបាន covariance និង correlation របស់សមីការ​ទាំងនោះ។ `np.cov` នឹងផ្តល់ឲ្យយើងនូវអ្វីដែលហៅថា **covariance matrix** ដែលជាការពង្រីកនៃ covariance ទៅនឹងអថេរជាច្រើន។ ធាតុ $M_{ij}$ នៃម៉ាទ្រីស covariance $M$ គឺជាការតភ្ជាប់រវាងអថេរ $X_i$ និង $X_j$ ផ្ទុកក្នុងចំណូល ដោយតម្លៃគន្លងបន្ទាត់ $M_{ii}$ គឺជាអាចប្រើបាននៃ $X_i$។ ដូចគ្នា `np.corrcoef` នឹងផ្តល់ឲ្យយើងនូវ **ម៉ាទ្រីស correlation** ។


```python
print(f"Covariance matrix:\n{np.cov(heights, salaries)}")
print(f"Covariance = {np.cov(heights, salaries)[0,1]}")
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

ការតភ្ជាប់ស្មើនឹង 1 មានន័យថាមាន​​**សម្ព័ន្ធបន្ទាត់**ខ្លាំងរវាងអថេរចម្លងពីរគ្នា។ យើងអាចមើលឃើញសម្ព័ន្ធបន្ទាត់នេះបានតាមរយៈការរូបតាងតម្លៃមួយប្រៀបធៀបទៅនឹងមួយផ្សេងទៀត៖


```python
plt.figure(figsize=(10,6))
plt.scatter(heights,salaries)
plt.tight_layout()
plt.show()
```

មកបើកមើលថាអ្វីដែលកើតឡើងបើទំនាក់ទំនងមិនមែនជាផ្លូវបន្ទាត់។ សូមស្រមៃថាក្រុមហ៊ុនរបស់យើងបានសម្រេចចិត្តលាក់ការពឹងផ្អែកផ្លូវបន្ទាត់ដែលច្បាស់លាស់រវាងជម្រោយកម្ពស់ និងប្រាក់ខែ ហើយបានបញ្ចូលករណីមិនមែនផ្លូវបន្ទាត់ក្នុងរូបមន្ត ដូចជា `sin`៖


```python
salaries = 1000+np.sin((heights-heights.min())/(heights.max()-heights.mean()))*100
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

ក្នុងករណីនេះ ការតភ្ជាប់គ្នាមានទំហំតិចបន្តិច ប៉ុន្តែវានៅតែខ្ពស់ជាងមធ្យម។ ឥឡូវនេះ ដើម្បីធ្វើអោយទំនាក់ទំនងកាន់តែអស្បថ យើងអាចចង់បន្ថែម ភាពចៃដន្យបន្ថែមមួយចំនួន ដោយបន្ថែមអថេរចៃដន្យមួយចំនួនទៅលើប្រាក់ខែ។ សូមមើលថាមើលមានអ្វីកើតឡើង៖


```python
salaries = 1000+np.sin((heights-heights.min())/(heights.max()-heights.mean()))*100+np.random.random(size=len(heights))*20-10
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

```python
plt.figure(figsize=(10,6))
plt.scatter(heights, salaries)
plt.tight_layout()
plt.show()
```

> តើអ្នកអាចគិតចិត្​យថាហេតុអ្វីបានជាចំណុចដូចជា​បានជួរតូចៗកាត់មានជួរដេកដូចនេះ?

យើងបានសង្កេតឃើញការតភ្ជាប់រវាងយុទ្ធសាស្ត្រដែលបានបង្កើតឡើងដោយសិប្បនិម្មិតដូចជាការប្រាក់និងអថេរដែលបានសង្កេត *កម្ពស់*។ យើងក៏ព្យាយាមមើលថាតើអថេរសង្កេតពីរ ដូចជា កម្ពស់ និងទម្ងន់ មានការ​តភ្ជាប់គ្នាផងដែរឬទេ៖


```python
np.corrcoef(df['Height'].ffill(),df['Weight'])
```

> ឧទាហរណ៍នេះបង្ហាញពីសារៈសំខាន់នៃ **ការរៀបចំទិន្នន័យ** និង **ការសម្អាត**។ គ្មានទិន្នន័យត្រឹមត្រូវ យើងមិនអាចគណនាអ្វីបានទេ។

យើងមកប្រើវិធីសាស្រ្ត `fillna` ដើម្បីបំពេញតម្លៃដែលខ្វះ ហើយគណនាការប្រៀបធៀប:
