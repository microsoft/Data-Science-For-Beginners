## Lengo

Katika somo hili, tumekuwa tukijadili dhana tofauti zinazohusiana na Sayansi ya Data. Tujaribu kugundua dhana zaidi zinazohusiana kwa kufanya **uchimbaji wa maandishi**. Tutaanza na maandishi kuhusu Sayansi ya Data, kutoa maneno muhimu kutoka kwake, kisha kujaribu kuonyesha matokeo kwa njia ya picha.

Kama maandishi, nitaitumia ukurasa kuhusu Sayansi ya Data kutoka Wikipedia:


```python
url = 'https://en.wikipedia.org/wiki/Data_science'
```

## Hatua ya 1: Kupata Takwimu

Hatua ya kwanza katika kila mchakato wa sayansi ya takwimu ni kupata takwimu. Tutatumia maktaba ya `requests` kufanya hivyo:


```python
import requests

# Define a custom header.
headers = {
    'User-Agent': 'DataScienceChallenge/1.0 (myemail@gmail.com)'
}

# Pass the headers into the get request
response = requests.get(url, headers=headers)

if response.status_code == 200:
    text = response.content.decode('utf-8')
    print(text[:1000])
else:
    print(f"Error: {response.status_code}")
```

## Hatua ya 2: Kubadilisha Data

Hatua inayofuata ni kubadilisha data kuwa katika muundo unaofaa kwa ajili ya usindikaji. Katika kesi yetu, tumepakua msimbo wa chanzo wa HTML kutoka ukurasa, na tunahitaji kuubadilisha kuwa maandishi ya kawaida.

Kuna njia nyingi za kufanya hivi. Tutatumia [BeautifulSoup](https://www.crummy.com/software/BeautifulSoup/), maktaba maarufu ya Python kwa uchambaji wa HTML. BeautifulSoup inatupa uwezo wa kulenga vipengele maalum vya HTML, hivyo tunaweza kuzingatia maudhui kuu ya makala kutoka Wikipedia na kupunguza baadhi ya menyu za urambazaji, bar zilizopo pembeni, miguso ya chini, na maudhui mengine yasiyohusiana (ingawa baadhi ya maandishi ya msingi yanaweza kubaki).


Kwanza, tunahitaji kusakinisha maktaba ya BeautifulSoup kwa uchambuzi wa HTML:


```python
import sys
!{sys.executable} -m pip install beautifulsoup4
```

```python
from bs4 import BeautifulSoup

# Parse the HTML content
soup = BeautifulSoup(text, 'html.parser')

# Extract only the main article content from Wikipedia
# Wikipedia uses 'mw-parser-output' class for the main article content
content = soup.find('div', class_='mw-parser-output')

def clean_wikipedia_content(content_node):
    """Remove common non-article elements from a Wikipedia content node."""
    # Strip jump links, navboxes, reference lists/superscripts, edit sections, TOC, sidebars, etc.
    selectors = [
        '.mw-jump-link',
        '.navbox',
        '.reflist',
        'sup.reference',
        '.mw-editsection',
        '.hatnote',
        '.metadata',
        '.infobox',
        '#toc',
        '.toc',
        '.sidebar',
    ]
    for selector in selectors:
        for el in content_node.select(selector):
            el.decompose()

if content:
    # Clean the content node to better approximate article text only.
    clean_wikipedia_content(content)
    text = content.get_text(separator=' ', strip=True)
    print(text[:1000])
else:
    print("Could not find main content. Using full page text.")
    text = soup.get_text(separator=' ', strip=True)
    print(text[:1000])
```

## Hatua ya 3: Kupata Uelewa

Hatua muhimu zaidi ni kugeuza data yetu kuwa aina fulani kutoka ambayo tunaweza kutoa uelewa. Kwa kesi yetu, tunataka kutoa maneno muhimu kutoka kwa maandishi, na kuona ni maneno gani muhimu zaidi.

Tutatumia maktaba ya Python inayoitwa [RAKE](https://github.com/aneesha/RAKE) kwa ajili ya kutoa maneno muhimu. Kwanza, wacha tuiweke maktaba hii kama haipo: 


```python
import sys
!{sys.executable} -m pip install nlp_rake
```

Kazi kuu inapatikana kutoka kwa kitu `Rake`, ambacho tunaweza kubinafsisha kwa kutumia vigezo baadhi. Katika kesi yetu, tutaweka urefu mdogo wa neno kuu kuwa wahusika 5, mara kidogo neno kuu linapotokea katika hati kuwa 3, na idadi kubwa ya maneno katika neno kuu - kuwa 2. Jisikie huru kucheza na thamani nyingine na kuangalia matokeo.


```python
import nlp_rake
extractor = nlp_rake.Rake(max_words=2,min_freq=3,min_chars=5)
res = extractor.apply(text)
res
```


Tulipata orodha ya maneno pamoja na kiwango kinachohusiana cha umuhimu. Kama unavyoweza kuona, fani muhimu zaidi, kama vile ujifunzaji wa mashine na data kubwa, zipo kwenye orodha katika nafasi za juu.

## Hatua ya 4: Kuona Matokeo Kwa Njia ya Mchoro

Watu wanaweza kufasiri data vizuri zaidi kwa njia ya kuona. Hivyo mara nyingi ina maana kuonyesha data ili kupata uelewa fulani. Tunaweza kutumia maktaba ya `matplotlib` katika Python kuchora usambazaji rahisi wa maneno muhimu pamoja na umuhimu wake:


```python
import matplotlib.pyplot as plt

def plot(pair_list):
    k,v = zip(*pair_list)
    plt.bar(range(len(k)),v)
    plt.xticks(range(len(k)),k,rotation='vertical')
    plt.show()

plot(res)
```

Hata hivyo, kuna njia bora zaidi ya kuona mara ngapi maneno yanatokea - kutumia **Word Cloud**. Tutahitaji kusanidi maktaba nyingine ili kuchora wingu la maneno kutoka kwenye orodha yetu ya maneno muhimu.


```python
!{sys.executable} -m pip install wordcloud
```

Kitu cha `WordCloud` kinahusika na kupokea ama maandishi ya asili, au orodha iliyohesabiwa awali ya maneno na mara ngapi yanapotokea, na kurudisha picha, ambayo inaweza kuonyeshwa kwa kutumia `matplotlib`:


```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

wc = WordCloud(background_color='white',width=800,height=600)
plt.figure(figsize=(15,7))
plt.imshow(wc.generate_from_frequencies({ k:v for k,v in res }))
```

Tunaweza pia kupitisha maandishi ya awali kwa `WordCloud` - tuone kama tunaweza kupata matokeo yanayofanana:


```python
plt.figure(figsize=(15,7))
plt.imshow(wc.generate(text))
```
