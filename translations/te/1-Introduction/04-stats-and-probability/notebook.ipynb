## విశ్వసనీయత అంతर्वిలయాలు

ఇప్పుడు బేస్‌బాల్ ఆటగాళ్ళ బరువు మరియు ఎత్తుల కోసం విశ్వసనీయత అంతర్వილయాలను లెక్కించుకుందాం. మనం [ఈ stackoverflow చర్చలోని కోడ్](https://stackoverflow.com/questions/15033511/compute-a-confidence-interval-from-sample-data) ను ఉపయోగించుకుంటాము:


```python
import scipy.stats

def mean_confidence_interval(data, confidence=0.95):
    a = 1.0 * np.array(data)
    n = len(a)
    m, se = np.mean(a), scipy.stats.sem(a)
    h = se * scipy.stats.t.ppf((1 + confidence) / 2., n-1)
    return m, h

for p in [0.85, 0.9, 0.95]:
    m, h = mean_confidence_interval(df['Weight'].ffill(),p)
    print(f"p={p:.2f}, mean = {m:.2f} ± {h:.2f}")
```

## హైపోథసిస్ పరీక్ష

మన బేస్ బాల్ ఆడేవార డేటాసెట్‌లో వివిధ పాత్రలను పరిశీలిద్దాం:


```python
df.groupby('Role').agg({ 'Weight' : 'mean', 'Height' : 'mean', 'Age' : 'count'}).rename(columns={ 'Age' : 'Count'})
```

మనం మొదటి బేస్‌మెన్‌లు రెండవ బేస్‌మెన్‌ల కంటే ఎత్తైనవారా అనే ఊహను పరీక్షిద్దాం. దీనిని చేయడానికి సరళమైన మార్గం విశ్వాస యంత్రాల్ని పరీక్షించడం:


```python
for p in [0.85,0.9,0.95]:
    m1, h1 = mean_confidence_interval(df.loc[df['Role']=='First_Baseman',['Height']],p)
    m2, h2 = mean_confidence_interval(df.loc[df['Role']=='Second_Baseman',['Height']],p)
    print(f'Conf={p:.2f}, 1st basemen height: {m1-h1[0]:.2f}..{m1+h1[0]:.2f}, 2nd basemen height: {m2-h2[0]:.2f}..{m2+h2[0]:.2f}')
```

మేము ఇంటర్వాల్స్ అతుకులుపోనట్లు కనుగొంటాం.

ఒక గణాంకంగా మరింత సరి అయిన విధానం అనే హైపోతీసిస్ ను సాక్ష్యపరచడానికి **స్టూడెంట్ t-టెస్ట్** ను ఉపయోగించడం.


```python
from scipy.stats import ttest_ind

tval, pval = ttest_ind(df.loc[df['Role']=='First_Baseman',['Height']], df.loc[df['Role']=='Second_Baseman',['Height']],equal_var=False)
print(f"T-value = {tval[0]:.2f}\nP-value: {pval[0]}")
```

`ttest_ind` ఫంక్షన్ తిరుగు ఇచ్చే రెండు విలువలు:
* p-విలువను రెండు వితరణల సగటు ఒకే విధంగా ఉండే అవకాశంగా పరిగణించవచ్చు. మన కేసులో, ఇది చాలా తక్కువగా ఉంది, అంటే ఫస్ట్ బేస్మేన్లు ఎక్కువ ఎత్తుగలవారని బలమైన సాక్ష్యం ఉంది.
* t-విలువు అనేది సగటు తేడా సాధారణీకరించిన మధ్యంతర విలువు, ఇది t-టెస్ట్‌లో ఉపయోగిస్తారు, మరియు ఇచ్చిన విశ్వసనీయత విలువ కోసం ఒక సరిహద్దు విలువతో పోల్చబడుతుంది.


## సెంట్రల్ లిమిట్ థియరీతో సాధారణ పంపిణీని అనుకరించడం

పైథాన్ లోని ప్స్యూడో-యాదృచ్ఛిక జనరేటర్ మనకు ఒక సమమైన పంపిణీని అందించే విధంగా రూపొందించబడింది. సాధారణ పంపిణీ కోసం ఒక జనరేటర్ సృష్టించాలనుకుంటే, మనం సెంట్రల్ లిమిట్ థియరీను ఉపయోగించవచ్చు. సాధారణ పంపిణీ విలువను పొందడానికి, మనం సమమైనంగా ఉత్పన్నమైన నమూనా యొక్క సగటును లెక్కించాలి.


```python
def normal_random(sample_size=100):
    sample = [random.uniform(0,1) for _ in range(sample_size) ]
    return sum(sample)/sample_size

sample = [normal_random() for _ in range(100)]
plt.figure(figsize=(10,6))
plt.hist(sample)
plt.tight_layout()
plt.show()
```

## సంబంధం మరియు చెడ్డ బేస్‌బాల్ కార్పొరేషన్

సంబంధం మనకు డేటా వరుసల మధ్య సంబంధాలను కనుగొనడానికి సహాయపడుతుంది. మన యొక్క ఆట మాదిరిలో, ఒక చెడ్డ బేస్‌బాల్ కార్పొరేషన్ ఉందని భావిద్దాం, వారు తమ ఆటగాళ్లకు వారి ఎత్తు ఆధారంగా పే చేస్తుంది - ఆటగాడు ఎత్తైనట్టే, అతని/ఆమెకు ఎక్కువ డబ్బు అందుతుంది. ఒక ఆధార జీతం $1000 ఉంది అని, మరియు ఎత్తు ఆధారంగా $0 నుండి $100 అదనపు బోనస్ ఉంటుంది అని అనుకుందాం. మనం MLB లోని నిజమైన ఆటగాళ్లను తీసుకుని, వారి ఊహాత్మక జీతాలను లెక్కిస్తాము:


```python
heights = df['Height'].ffill()
salaries = 1000+(heights-heights.min())/(heights.max()-heights.mean())*100
print(list(zip(heights, salaries))[:10])
```

ఇప్పుడు ఆ సీక్వెన్సుల యొక్క సహవిభేదం మరియు సంబంధాన్ని గణించుకుందాం. `np.cov` మనకు ఒక所谓మైన **సహవిభేద మేట్రిక్స్** ఉత్పత్తి చేస్తుంది, ఇది బహు చరకాలను కలిపిన సహవిభేదానికి విస్తరణ. సహవిభేద మేట్రిక్స్ $M$ లోని అంశం $M_{ij}$ అనేది ఇన్‌పుట్ చరకాల $X_i$ మరియు $X_j$ మధ్య సహవిభేదం, మరియు ద్రవ్యరాశి విలువ $M_{ii}$ అనేది $X_{i}$ యొక్క విభిన్నత. అదేవిధంగా, `np.corrcoef` మనకు **సంబంధ మేట్రిక్స్** ఇస్తుంది.


```python
print(f"Covariance matrix:\n{np.cov(heights, salaries)}")
print(f"Covariance = {np.cov(heights, salaries)[0,1]}")
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

1 సమానమైన సహసంబంధం అంటే ఇద్దరు చరాలు మధ్య బలమైన **రేఖీయ సంబంధం** ఉందని అర్థం. ఒక విలువను против ఎదుట మరోటి ప్లాట్ చేసి మేము ఈ రేఖీయ సంబంధాన్ని దృశ్యంగా చూడవచ్చు:


```python
plt.figure(figsize=(10,6))
plt.scatter(heights,salaries)
plt.tight_layout()
plt.show()
```

సంబంధం రేఖీయమైనది కాకపోతే ఏం జరుగుతుందో చూద్దాం. మన సంస్థ ఎత్తుల మరియు జీతాల మధ్య స్పష్టమైన రేఖీయ ఆధారాన్ని దాచాలని నిర్ణయించుకుంది మరియు సూత్రంలో కొంత రేఖీయత లేని వ్యవధిని పరిచయం చేసింది, ఉదాహరణకు `sin` వంటి:  


```python
salaries = 1000+np.sin((heights-heights.min())/(heights.max()-heights.mean()))*100
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

ఈ సందర్భంలో, సంబంధం కొంచెం తక్కువగా ఉన్నా, అది ఇంకా చాలా ఎక్కువగా ఉంటుంది. ఇప్పుడు, సంబంధాన్ని మరింత తక్కువగా చూపించడానికి, జీతానికి కొంత అదనపు యాదృచ్ఛికాన్ని చేర్చాలని మేము అనుకోవచ్చు. ఏమవుతుందని చూద్దాం:


```python
salaries = 1000+np.sin((heights-heights.min())/(heights.max()-heights.mean()))*100+np.random.random(size=len(heights))*20-10
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

```python
plt.figure(figsize=(10,6))
plt.scatter(heights, salaries)
plt.tight_layout()
plt.show()
```

> మీరు ఎందుకు ఈ మచ్చలు నిలువు గీతలుగా వరుసగా నిలుస్తాయనే అంచనా వేశారా?

జీతం వంటి కృత్రిమంగా రూపొంచబడిన సങ്കల్పం మరియు గమనించిన వేరియబులైన *ఎత్తు* మధ్య సంబంధాన్ని మేము గమనించాము. అలాగే, ఎత్తు మరియు బరువు వంటి గమనించిన రెండు వేరియబుల్స్ కూడా పరస్పరం సంబంధం ఉందా అని కూడా చూద్దాం:


```python
np.corrcoef(df['Height'].ffill(),df['Weight'])
```

దురదృష్టవశాత్తూ, మాకు ఎటువంటి ఫలితాలు రాలేదు - కేవలం కొంత వింత `nan` విలువలు మాత్రమే ఉన్నాయి. మా సిరీస్‌లో కొన్ని విలువలు నిర్వచించబడకపోవడం కారణంగా, అవి `nan` గా సూచించబడ్డాయి, ఇది ఆపరేషన్ ఫలితం కూడా నిర్వచించబడదని కారణమవుతుంది. మ్యాట్రిక్స్‌ను చూసినప్పుడు మనం గమనించగలము, `Weight` అనేది సమస్యాత్మక కాలమ్, ఎందుకంటే `Height` విలువల మధ్య స్వీయ-సంబంధం లెక్కించబడింది.

> ఈ ఉదాహరణ **డేటా సిద్ధత** మరియు **శుభ్రపరచడం** యొక్క ప్రాముఖ్యతను చూపిస్తుంది. సరైన డేటా లేకుండా మనం ఏదీ లెక్కించలేం.

మిస్సింగ్ విలువలను పూరించడానికి `fillna` పద్ధతిని ఉపయోగించి, సంబంధం లెక్కించుకుందాం: 


```python
np.corrcoef(df['Height'].ffill(), df['Weight'])
```

నిజానికి సంబంధం ఉంటుందిఅయితే, మన కళాకృతితో చేసిన ఉదాహరణ లో ఉండేలాగా బలమైనది కాదు. నిజానికి, ఒక విలువను మరొక విలువతో స్కాటర్ ప్లాట్ చూస్తే, సంబంధం చాలా తక్కువగా కనిపిస్తుంది:
