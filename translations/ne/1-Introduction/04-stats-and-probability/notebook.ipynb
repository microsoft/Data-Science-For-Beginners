## र्यान्डम भेरिएबलहरू र वितरणहरू
आउनुहोस् ० देखि ९ सम्मको एउटा समान वितरणबाट ३० मानहरूको नमुना तान्नेबाट सुरु गरौं। हामी औसत र भिन्नता पनि गणना गर्नेछौं।

```python
wrong_sample = np.random.rand(1000)*2*std+mean-std
plt.figure(figsize=(10,6))
plt.hist(wrong_sample)
plt.tight_layout()
plt.show()
```

## आत्मविश्वास अन्तराल

अब हामी बेसबल खेलाडीहरूको तौल र उचाइका लागि आत्मविश्वास अन्तराल गणना गर्नेछौं। हामी यो कोड प्रयोग गर्नेछौं [यस stackoverflow छलफलबाट](https://stackoverflow.com/questions/15033511/compute-a-confidence-interval-from-sample-data):


```python
import scipy.stats

def mean_confidence_interval(data, confidence=0.95):
    a = 1.0 * np.array(data)
    n = len(a)
    m, se = np.mean(a), scipy.stats.sem(a)
    h = se * scipy.stats.t.ppf((1 + confidence) / 2., n-1)
    return m, h

for p in [0.85, 0.9, 0.95]:
    m, h = mean_confidence_interval(df['Weight'].ffill(),p)
    print(f"p={p:.2f}, mean = {m:.2f} ± {h:.2f}")
```

## परिकल्पना परीक्षण

हाम्रा बेसबल खेलाडी डाटासेटमा फरक भुमिकाहरू अन्वेषण गरौं:


```python
df.groupby('Role').agg({ 'Weight' : 'mean', 'Height' : 'mean', 'Age' : 'count'}).rename(columns={ 'Age' : 'Count'})
```

पहिलो बेसम्यान दोस्रो बेसम्यानहरू भन्दा अग्लो हुन्छन् भन्ने अनुमानको परीक्षण गरौं। यसको लागि सबैभन्दा सरल तरिका भनेको आत्मविश्वास अन्तरालहरूको परीक्षण गर्नु हो:


```python
for p in [0.85,0.9,0.95]:
    m1, h1 = mean_confidence_interval(df.loc[df['Role']=='First_Baseman',['Height']],p)
    m2, h2 = mean_confidence_interval(df.loc[df['Role']=='Second_Baseman',['Height']],p)
    print(f'Conf={p:.2f}, 1st basemen height: {m1-h1[0]:.2f}..{m1+h1[0]:.2f}, 2nd basemen height: {m2-h2[0]:.2f}..{m2+h2[0]:.2f}')
```

हामी देख्न सक्छौं कि अन्तरालहरू ओभरल्याप हुँदैनन्।

परिकल्पनालाई प्रमाणित गर्ने एउटा सांख्यिकीय रूपमा सही तरिका भनेको **स्टुडेन्ट टी-टेस्ट** प्रयोग गर्नु हो:


```python
from scipy.stats import ttest_ind

tval, pval = ttest_ind(df.loc[df['Role']=='First_Baseman',['Height']], df.loc[df['Role']=='Second_Baseman',['Height']],equal_var=False)
print(f"T-value = {tval[0]:.2f}\nP-value: {pval[0]}")
```

`ttest_ind` कार्यले फर्काइने दुई मानहरू हुन्:
* p-मूल्य दुई वितरणहरूको एउटै औसत हुने सम्भावना मान्न सकिन्छ। हाम्रो अवस्थामा, यो धेरै कम छ, जसको अर्थ छ कि प्रथम बेसम्यानहरू उच्च हुनुको लागि मजबूत प्रमाण छ।
* t-मूल्य सामान्यीकृत औसत भिन्नताको मध्यवर्ती मान हो जुन t-परीक्षणमा प्रयोग गरिन्छ, र यसलाई दिइएको विश्वस्तताका लागि थ्रेसहोल्ड मानसँग तुलना गरिन्छ।


## केन्द्रीय सीमा प्रमेयको साथ साधारण वितरणको अनुकरण

पाइथनमा स्यूडो-र्याण्डम जेनरेटरले हामीलाई एकसमान वितरण दिन डिजाइन गरिएको छ। यदि हामीलाई साधारण वितरणको लागि जेनरेटर बनाउनु छ भने, हामी केन्द्रीय सीमा प्रमेय प्रयोग गर्न सक्छौं। सामान्य रूपमा वितरण गरिएको मान प्राप्त गर्न हामीले एकसमान-उत्पन्न नमुनाको माध्य गणना गर्नेछौं।


```python
def normal_random(sample_size=100):
    sample = [random.uniform(0,1) for _ in range(sample_size) ]
    return sum(sample)/sample_size

sample = [normal_random() for _ in range(100)]
plt.figure(figsize=(10,6))
plt.hist(sample)
plt.tight_layout()
plt.show()
```

## सहसम्बन्ध र दुष्ट बेसबल कर्पोरेशन

सहसम्बन्धले हामीलाई डाटा अनुक्रमहरूबीच सम्बन्ध पत्ता लगाउन मद्दत गर्दछ। हाम्रो सानो उदाहरणमा, मानौं त्यहाँ एक दुष्ट बेसबल कर्पोरेशन छ जसले आफ्ना खेलाडीलाई तिनीहरूको उचाइ अनुसार तलब दिन्छ - जस खेलाडी उचो हुन्छ, उसलाई बढी पैसा प्राप्त हुन्छ। मानौं आधारभूत तलब $१००० छ, र उचाइअनुसार $० देखि $१०० सम्म अतिरिक्त बोनस पाइन्छ। हामी MLB का वास्तविक खेलाडीहरू लिनेछौं र तिनीहरूको काल्पनिक तलब गणना गर्नेछौं:


```python
heights = df['Height'].ffill()
salaries = 1000+(heights-heights.min())/(heights.max()-heights.mean())*100
print(list(zip(heights, salaries))[:10])
```

अब ती अनुक्रमहरूको सहसम्बन्ध र सहसञ्चिका गणना गरौं। `np.cov` ले हामीलाई तथाकथित **सहसञ्चिका म्याट्रिक्स** दिनेछ, जुन धेरै चरहरूमा सहसञ्चिकाको विस्तार हो। सहसञ्चिका म्याट्रिक्स $M$ को तत्त्व $M_{ij}$ ले इनपुट चरहरू $X_i$ र $X_j$ बीचको सहसम्बन्ध जनाउँछ, र विकर्ण मानहरू $M_{ii}$ $X_{i}$ को विचरण हो। त्यसै गरी, `np.corrcoef` ले हामीलाई **सहसम्बन्ध म्याट्रिक्स** दिनेछ।


```python
print(f"Covariance matrix:\n{np.cov(heights, salaries)}")
print(f"Covariance = {np.cov(heights, salaries)[0,1]}")
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

१ को सहसम्बन्धले जनाउँछ कि दुई चलहरूबीच एउटा बलियो **रेखीय सम्बन्ध** छ। हामी एउटा मानलाई अर्कोको विरुद्ध प्लट गरेर रेखीय सम्बन्धलाई दृश्य रूपमा देख्न सक्छौं:


```python
plt.figure(figsize=(10,6))
plt.scatter(heights,salaries)
plt.tight_layout()
plt.show()
```

यो सम्बन्ध रेखीय नभए के होला भनेर हेरौं। मानौं हाम्रो कम्पनीले उचाइ र तलब बीचको स्पष्ट रेखीय निर्भरता लुकाउने निर्णय गर्यो, र सूत्रमा केहि गैर-रेखीयता, जस्तै `sin`, प्रस्तुत गर्यो:


```python
salaries = 1000+np.sin((heights-heights.min())/(heights.max()-heights.mean()))*100
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

यस अवस्थामा, सहसम्बन्ध अलि सानो छ, तर यो अझै पनि निकै उच्च छ। अब, सम्बन्धलाई अझ कम स्पष्ट बनाउनका लागि, हामी तलबमा केही अतिरिक्त यादृच्छिकता थप्न कुनै यादृच्छिक भेरिएबल थप्न चाहन सक्छौं। हेर्नुहोस् के हुन्छ:


```python
salaries = 1000+np.sin((heights-heights.min())/(heights.max()-heights.mean()))*100+np.random.random(size=len(heights))*20-10
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

```python
plt.figure(figsize=(10,6))
plt.scatter(heights, salaries)
plt.tight_layout()
plt.show()
```

> तपाईं अनुमान लगाउन सक्नुहुन्छ किन डटहरू यसरी ठाडो लाइनहरूमा सिधा हुन्छन्?

हामीले तलबजस्ता कृत्रिम रूपमा तयार गरिएको अवधारणा र अवलोकित चर *उचाइ* बीचको सम्बन्धलाई अवलोकन गरेका छौं। अब हामीले हेर्न सक्छौं कि दुई अवलोकित चरहरू, जस्तै उचाइ र तौल, पनि सम्बन्धित छन् कि छैनन्:


```python
np.corrcoef(df['Height'].ffill(),df['Weight'])
```

आउनुहोस् `fillna` विधि प्रयोग गरी हराएको मानहरू भर्नुहोस्, र सहसम्बन्ध गणना गरौं:

वास्तवमा एक सम्बन्ध छ, तर हाम्रो कृत्रिम उदाहरणजस्तै कडा छैन। वास्तवमै, यदि हामी एउटा मानलाई अर्को मानसँग तुलना गर्ने स्क्याटर प्लट हेर्छौं भने सम्बन्ध धेरै कम स्पष्ट हुन्छ:
