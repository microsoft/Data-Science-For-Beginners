## நம்பிக்கை இடைவெளிகள்

இப்போது நாம் பேஸ்பால் வீரர்களின் எடைகள் மற்றும் உயரங்களுக்கு நம்பிக்கை இடைவெளிகளை கணக்கிடுவோம். நாம் இந்தக் கோடை [இந்த stackoverflow விவாதத்திலிருந்து](https://stackoverflow.com/questions/15033511/compute-a-confidence-interval-from-sample-data) பயன்படுத்துகிறோம்:


```python
import scipy.stats

def mean_confidence_interval(data, confidence=0.95):
    a = 1.0 * np.array(data)
    n = len(a)
    m, se = np.mean(a), scipy.stats.sem(a)
    h = se * scipy.stats.t.ppf((1 + confidence) / 2., n-1)
    return m, h

for p in [0.85, 0.9, 0.95]:
    m, h = mean_confidence_interval(df['Weight'].ffill(),p)
    print(f"p={p:.2f}, mean = {m:.2f} ± {h:.2f}")
```

## ஊகசோதனை

நாம் எமது பேஸ்பால் வீரர் தரவுத்தொகுப்பில் உள்ள வெவ்வேறு பாத்திரங்களை ஆராய்ந்து பார்க்கலாம்:


```python
df.groupby('Role').agg({ 'Weight' : 'mean', 'Height' : 'mean', 'Age' : 'count'}).rename(columns={ 'Age' : 'Count'})
```

முதன்மை பேஸ்மேன்கள் இரண்டாவது பேஸ்மேன்களுக்கு விட உயரமாக இருப்பதாக உள்ள கருத்தை பரிசோதிப்போம். இதைச் செய்வதற்கான எளிதான வழி நம்பிக்கைக் கோட்டுகளை தேர்வு செய்வதே ஆகும்:


```python
for p in [0.85,0.9,0.95]:
    m1, h1 = mean_confidence_interval(df.loc[df['Role']=='First_Baseman',['Height']],p)
    m2, h2 = mean_confidence_interval(df.loc[df['Role']=='Second_Baseman',['Height']],p)
    print(f'Conf={p:.2f}, 1st basemen height: {m1-h1[0]:.2f}..{m1+h1[0]:.2f}, 2nd basemen height: {m2-h2[0]:.2f}..{m2+h2[0]:.2f}')
```

இடைவேளைகள் ஒரே நேரத்தில் இல்லாததை நாம் பார்க்க முடிகிறது.

கருத்தை பறைசாற்றுவதற்கு புள்ளியியல் ரீதியாக சரியான வழி **ஸ்டூடென்ட் t-சோதனை** பயன்படுத்துவது ஆகும்:


```python
from scipy.stats import ttest_ind

tval, pval = ttest_ind(df.loc[df['Role']=='First_Baseman',['Height']], df.loc[df['Role']=='Second_Baseman',['Height']],equal_var=False)
print(f"T-value = {tval[0]:.2f}\nP-value: {pval[0]}")
```

`ttest_ind` செயல்பாடு வழங்கும் இரண்டு மதிப்புகள்:
* p-மதிப்பு என்பது இரண்டு பகிர்வுகளின் சராசரி ஒரே மாதிரியானவை என்பதற்கான சாத்தியமாக கருதப்படலாம். எங்கள் விவகாரத்தில், இது மிகவும் குறைவாக உள்ளது, இது முதல் பேஸ்மேன் உயரமாக இருப்பதற்கான பலமான ஆதாரத்தை குறிக்கின்றது.
* t-மதிப்பு என்பது t-சோதனையில் பயன்படுத்தப்படும் சாதாரணப்படுத்தப்பட்ட சராசரி வேறுபாட்டின் நடுநிலை மதிப்பாகும், மேலும் இது கொடுக்கப்பட்ட நம்பிக்கை மதிப்பிற்கான எல்லை மதிப்புடன் ஒப்பிடப்படுகிறது.


## மைய வரம்பு கோட்பாட்டுடன் சாதாரண பங்கீட்டைக் கண்காணித்தல்

பைத்தானில் உள்ள புது-சீரற்ற உருவாக்கி ஒரு ஒரே மாதிரிப் பங்கீடு வழங்க உருவாக்கப்பட்டது. நாம் சாதாரண பங்கீட்டுக்கான உருவாக்கியை உருவாக்க விரும்பினால், மைய வரம்பு கோட்பாட்டைப் பயன்படுத்தலாம். சாதாரணமாக பங்கீட்டுப்பெறும் மதிப்பை பெற நாம் ஒட்டுமொத்தமாக உருவாக்கப்பட்ட மாதிரியின் சராசரியை கணக்கிடுவோம்.


```python
def normal_random(sample_size=100):
    sample = [random.uniform(0,1) for _ in range(sample_size) ]
    return sum(sample)/sample_size

sample = [normal_random() for _ in range(100)]
plt.figure(figsize=(10,6))
plt.hist(sample)
plt.tight_layout()
plt.show()
```

## தொடர்பு மற்றும் தீய பேஸ்பால் கார்ப்பரேஷன்

தொடர்பு மூலம் நாம் தரவு தொடர்களுக்கு இடையிலான உறவுகளை கண்டறிய முடியும். நமது விளையாட்டு உதாரணத்தில், ஒரு தீய பேஸ்பால் கம்பெனி தங்கள் வீரர்களுக்கு அவர்களின் உயரத்துக்கு ஏற்ப சம்பளம் கொடுப்பதாக நினைக்கலாம் - உயரமானவர் વધારે பணம் பெறுவார். அடிப்படையான சம்பளம் $1000, மேலும் உயரம் அடிப்படையில் $0 முதல் $100 வரை ஒரு கூடுதல் போனஸ் உள்ளது என எடுத்துக்கொள்வோம். நாங்கள் MLB-இல் இருந்து உண்மையான வீரர்களை எடுத்துக் கொண்டு, அவர்களது கற்பனை சம்பளங்களை கணக்கிடுவோம்:


```python
heights = df['Height'].ffill()
salaries = 1000+(heights-heights.min())/(heights.max()-heights.mean())*100
print(list(zip(heights, salaries))[:10])
```

இப்போது அந்த தொடர்களின் கூட்டாற்பாட்டையும் தொடர்பட்டுக் கூர்தலையும் கணக்கிடுவோம். `np.cov`என்பது நமக்கு **கூட்டாற்பாட்டு அட்டவணையை** வழங்கும், இது பல மாறிலிகளுக்கு கூட்டாற்பாட்டின் விரிவாக்கம் ஆகும். கூட்டாற்பாட்டு அட்டவணை $M$ இன் கூறு $M_{ij}$ என்பது உள்ளீட்டு மாறிலிகள் $X_i$ மற்றும் $X_j$ ஆகியவற்றுக்கிடையேயான தொடர்பு ஆகும், மேலும் ஒருமுக எழுத்து வடிவிலுள்ள मानங்கள் $M_{ii}$ என்பது $X_{i}$ இன் பரவலாகும். அதேபோல், `np.corrcoef` நமக்கு **தொடர்புச் சுருக்க அட்டவணையை** வழங்கும்.


```python
print(f"Covariance matrix:\n{np.cov(heights, salaries)}")
print(f"Covariance = {np.cov(heights, salaries)[0,1]}")
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

1 என்ற தொடர்பு என்பது இரண்டு மாறிலிகள் இடையில் ஒரு வலுவான **எழுத்து சார்ந்த உறவு** இருப்பதை குறிக்கும். ஒருவருடைய மதிப்பை மற்றவர்மீது வரைதல் மூலம் எழுத்து சார்ந்த உறவைக் காணலாம்:


```python
plt.figure(figsize=(10,6))
plt.scatter(heights,salaries)
plt.tight_layout()
plt.show()
```

தொடர்பு நேரியல் அல்லாதது என்றால் என்ன நடக்கும் என்று பார்ப்போம். நமது நிறுவனமானது உயரங்களுக்கும் சம்பளங்களுக்குமான தெளிவான நேரியல் சார்பை மறைக்க முடிவு செய்தது, மேலும் சூத்திரத்தில் `sin` போன்ற ஒரு தவிர்க்க முடியாத மாறுபாட்டை அறிமுகப்படுத்தியது என்று கருதுக: 


```python
salaries = 1000+np.sin((heights-heights.min())/(heights.max()-heights.mean()))*100
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

இந்த நிலையில், தொடர்பு சிறிது குறைவாக உள்ளது, ஆனால் அது இன்னும் மிகவும் உயர்ந்ததே ஆகும். இப்போது, அந்த உறவை இன்னும் குறைவாகத் தெரியப்படுத்த, சம்பளத்திற்கு சில கூடுதல் சீரில்லாத தன்மைகளைச் சேர்க்க விரும்பலாம். என்ன நடக்கும் என்று பார்ப்போம்:


```python
salaries = 1000+np.sin((heights-heights.min())/(heights.max()-heights.mean()))*100+np.random.random(size=len(heights))*20-10
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

```python
plt.figure(figsize=(10,6))
plt.scatter(heights, salaries)
plt.tight_layout()
plt.show()
```

> இந்தப் புள்ளிகள் இப்படி செங்குத்து கோடுகளாக வரிசைப்படுத்தப்படுவது ஏன் என்று நீங்கள் கணிக்க முடியுமா?

சம்பளம் போன்ற செயற்கையாக வடிவமைக்கப்பட்ட கருத்து மற்றும் நோக்கப்பட்ட மாறிலி *உடலளவு* இடையேயான தொடர்பை நாம் கவனித்துள்ளோம். உயரம் மற்றும் எடை போன்ற இரண்டு நோக்கப்பட்ட மாறிலிகளும் தொடர்புடையவையா என்று பார்ப்போம்:


```python
np.corrcoef(df['Height'].ffill(),df['Weight'])
```

துரதிருஷ்டவசமாக, எங்களுக்கு எந்த முடிவுகளும் கிடைக்கவில்லை - சில அசாதாரண `nan` மதிப்புகள் மட்டும் கிடைத்தன. இது எமது வரிசை மதிப்புகளில் சில வரையறுக்கப்படாதவை, `nan` எனக் காணப்படுவது காரணமாக, செயல்பாட்டின் முடிவு கூட வரையறுக்கப்படாததாக மாறுகிறது. அட்டவணையைப் பார்க்கும்போது `Weight` என்பது பிரச்சனையான பத்தி என்பதை நாங்கள் காணலாம், ஏனெனில் `Height` மதிப்புக்கிடையேயான சுய தொடர்பு கணக்கிடப்பட்டுள்ளது.

> இந்த உதாரணம் **தரவு முன்னோட்டம்** மற்றும் **தெளிவேற்றத் திறன்** என்பவற்றின் முக்கியத்துவத்தை காட்டுகிறது. சரியான தரவின்றி எதையும் கணக்கிட முடியாது.

`fillna` முறையை பயன்படுத்தி இழந்த மதிப்புகளை நிரப்பி, தொடர்பை கணக்கிடுவோம்:


```python
np.corrcoef(df['Height'].ffill(), df['Weight'])
```

உண்மையில் ஒரு தொடர்பு உள்ளது, ஆனால் எங்கள் செயற்கை எடுத்துக்காட்டில் உள்ளதைப் போன்ற பலத்த தொடர்பு அல்ல. உண்மையில், நாம் ஒரு மதிப்பை மற்றொன்றுடன் ஒப்பிடும் புள்ளி வரைபடத்தைப் பார்த்தால், அந்த தொடர்பு மிகவும் தெளிவாக இருக்காது:
