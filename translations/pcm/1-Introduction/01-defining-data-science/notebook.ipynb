## Goal

As text, I go use di page wey dey for Data Science from Wikipedia:


```python
url = 'https://en.wikipedia.org/wiki/Data_science'
```

## Step 1: Getting the Data

First step for every data science process na to get the data. We go use `requests` library do am:


```python
import requests

# Define a custom header.
headers = {
    'User-Agent': 'DataScienceChallenge/1.0 (myemail@gmail.com)'
}

# Pass the headers into the get request
response = requests.get(url, headers=headers)

if response.status_code == 200:
    text = response.content.decode('utf-8')
    print(text[:1000])
else:
    print(f"Error: {response.status_code}")
```

## Step 2: Transfom di Data

Di next step na to change di data to the form wey fit make processing easy. For our case, we don download HTML source code from di page, an we need change am into plain text.

Plenty ways dey we fit do dis. We go use [BeautifulSoup](https://www.crummy.com/software/BeautifulSoup/), one correct Python library wey dem dey use parse HTML. BeautifulSoup dey allow us to select correct HTML elements, so we fit focus on main article content from Wikipedia and reduce some navigation menus, sidebars, footers, and other content wey no too important (though some boilerplate text fit still dey).


Fes, we need to install the BeautifulSoup library for HTML parsing:


```python
import sys
!{sys.executable} -m pip install beautifulsoup4
```

```python
from bs4 import BeautifulSoup

# Parse the HTML content
soup = BeautifulSoup(text, 'html.parser')

# Extract only the main article content from Wikipedia
# Wikipedia uses 'mw-parser-output' class for the main article content
content = soup.find('div', class_='mw-parser-output')

def clean_wikipedia_content(content_node):
    """Remove common non-article elements from a Wikipedia content node."""
    # Strip jump links, navboxes, reference lists/superscripts, edit sections, TOC, sidebars, etc.
    selectors = [
        '.mw-jump-link',
        '.navbox',
        '.reflist',
        'sup.reference',
        '.mw-editsection',
        '.hatnote',
        '.metadata',
        '.infobox',
        '#toc',
        '.toc',
        '.sidebar',
    ]
    for selector in selectors:
        for el in content_node.select(selector):
            el.decompose()

if content:
    # Clean the content node to better approximate article text only.
    clean_wikipedia_content(content)
    text = content.get_text(separator=' ', strip=True)
    print(text[:1000])
else:
    print("Could not find main content. Using full page text.")
    text = soup.get_text(separator=' ', strip=True)
    print(text[:1000])
```

## Step 3: Get Insights

Di most important step na to change our data into one kain form wey we fit use draw insight. For our case, we wan comot keywords from di text, and see which keywords dey more meaningful.

We go use Python library wey dem dey call [RAKE](https://github.com/aneesha/RAKE) for keyword extraction. First, make we install dis library in case e never dey: 


```python
import sys
!{sys.executable} -m pip install nlp_rake
```

Di main wok sabi dey available from `Rake` object, we fit customize am wit some parameters. For our case, we go set di minimum length of keyword to 5 characters, minimum frequency of keyword for di document to 3, and maximum number of words wey fit dey for keyword - to 2. Abeg feel free to play wit oda values and see di result.


```python
import nlp_rake
extractor = nlp_rake.Rake(max_words=2,min_freq=3,min_chars=5)
res = extractor.apply(text)
res
```


We don get list of terms plus di level of importance wey dey join dem. As you fit see, di most important disciplines, like machine learning and big data, dey for di top part of di list.

## Step 4: Visualize di Result

People fit understand di data beta wen na for visual form. So e dey make sense to visualize di data make we fit gather insight. We fit use `matplotlib` library for Python to plot simple distribution of di keywords plus their relevance:


```python
import matplotlib.pyplot as plt

def plot(pair_list):
    k,v = zip(*pair_list)
    plt.bar(range(len(k)),v)
    plt.xticks(range(len(k)),k,rotation='vertical')
    plt.show()

plot(res)
```

Bot, beta waya dey to show word frequencies - na **Word Cloud**. We go need install anoda library to fit plot word cloud from our keyword list.


```python
!{sys.executable} -m pip install wordcloud
```

`WordCloud` object dey responsible for to accept either original text, or pre-computed list of words with dia frequencies, and e go return image, we fit use `matplotlib` show:


```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

wc = WordCloud(background_color='white',width=800,height=600)
plt.figure(figsize=(15,7))
plt.imshow(wc.generate_from_frequencies({ k:v for k,v in res }))
```

We fit also chook di original text inside `WordCloud` - make we see if we fit get di same result:


```python
plt.figure(figsize=(15,7))
plt.imshow(wc.generate(text))
```

```python
wc.generate(text).to_file('images/ds_wordcloud.png')
```

You fit see sey dat word cloud don dey more impressive now, but e still get plenti noise (like unrelated words like `Retrieved on`). Plus, we get less keywords wey get two words, like *data scientist*, or *computer science*. Na becos RAKE algorithm dey do better work to select beta keywords from text.
