# ਚੈਲੈਂਜ: ਡੇਟਾ ਸਾਇੰਸ ਬਾਰੇ ਪਾਠ ਦਾ ਵਿਸ਼ਲੇਸ਼ਣ

ਇਸ ਉਦਾਹਰਨ ਵਿੱਚ, ਆਓ ਇੱਕ ਸਧਾਰਣ ਕਸਰਤ ਕਰੀਏ ਜੋ ਪਰੰਪਰਾਗਤ ਡੇਟਾ ਸਾਇੰਸ ਪ੍ਰਕਿਰਿਆ ਦੇ ਸਾਰੇ ਕਦਮਾਂ ਨੂੰ ਕਵਰ ਕਰਦੀ ਹੈ। ਤੁਹਾਡੇ ਕੋਲ ਕੋਈ ਕੋਡ ਲਿਖਣ ਦੀ ਲੋੜ ਨਹੀਂ ਹੈ, ਤੁਸੀਂ ਸਿਰਫ ਹੇਠਾਂ ਦਿੱਤੇ ਸੈੱਲਾਂ 'ਤੇ ਕਲਿੱਕ ਕਰਕੇ ਉਹਨਾਂ ਨੂੰ ਚਲਾ ਸਕਦੇ ਹੋ ਅਤੇ ਨਤੀਜਾ ਵੇਖ ਸਕਦੇ ਹੋ। ਇੱਕ ਚੈਲੈਂਜ ਵਜੋਂ, ਤੁਹਾਨੂੰ ਪ੍ਰੇਰਿਤ ਕੀਤਾ ਜਾਂਦਾ ਹੈ ਕਿ ਤੁਸੀਂ ਇਹ ਕੋਡ ਵੱਖ-ਵੱਖ ਡੇਟਾ ਨਾਲ ਕੋਸ਼ਿਸ਼ ਕਰੋ। 

## ਲਕੜੀ

ਇਸ ਪਾਠ ਵਿੱਚ, ਅਸੀਂ ਡੇਟਾ ਸਾਇੰਸ ਨਾਲ ਸਬੰਧਤ ਵੱਖ-ਵੱਖ ਧਾਰਣਾਵਾਂ ਬਾਰੇ ਚਰਚਾ ਕੀਤੀ ਹੈ। ਆਓ ਕੁਝ ਹੋਰ ਸਬੰਧਤ ਧਾਰਣਾਵਾਂ ਨੂੰ ਜਾਨਚਣ ਦੀ ਕੋਸ਼ਿਸ਼ ਕਰੀਏ **ਟੈਕਸਟ ਮਾਈਨਿੰਗ** ਕਰ ਕੇ। ਅਸੀਂ ਡੇਟਾ ਸਾਇੰਸ ਬਾਰੇ ਇੱਕ ਟੈਕਸਟ ਨਾਲ ਸ਼ੁਰੂ ਕਰਾਂਗੇ, ਉਸ ਵਿੱਚੋਂ ਮੁੱਖ ਸ਼ਬਦ ਕੱਢਾਂਗੇ, ਅਤੇ ਫਿਰ ਨਤੀਜੇ ਨੂੰ ਵਿਜੁਅਲਾਇਜ਼ ਕਰਨ ਦੀ ਕੋਸ਼ਿਸ਼ ਕਰਾਂਗੇ।

ਇੱਕ ਟੈਕਸਟ ਵਜੋਂ, ਮੈਂ ਵਿਕੀਪੀਡੀਆ ਤੋਂ ਡੇਟਾ ਸਾਇੰਸ ਬਾਰੇ ਪੰਨਾ ਵਰਤਾਂਗਾ:


```python
url = 'https://en.wikipedia.org/wiki/Data_science'
```

## ਕਦਮ 1: ਡਾਟਾ ਪ੍ਰਾਪਤ ਕਰਨਾ

ਹਰ ਡਾਟਾ ਸਾਇੰਸ ਪ੍ਰਕਿਰਿਆ ਦਾ ਪਹਿਲਾ ਕਦਮ ਡਾਟਾ ਪ੍ਰਾਪਤ ਕਰਨਾ ਹੁੰਦਾ ਹੈ। ਇਸ ਲਈ ਅਸੀਂ `requests` ਲਾਇਬ੍ਰੇਰੀ ਦੀ ਵਰਤੋਂ ਕਰਾਂਗੇ:


```python
import requests

# Define a custom header.
headers = {
    'User-Agent': 'DataScienceChallenge/1.0 (myemail@gmail.com)'
}

# Pass the headers into the get request
response = requests.get(url, headers=headers)

if response.status_code == 200:
    text = response.content.decode('utf-8')
    print(text[:1000])
else:
    print(f"Error: {response.status_code}")
```

## ਕਦਮ 2: ਡਾਟਾ ਦਾ ਰੂਪਾਂਤਰਣ

ਅਗਲਾ ਕਦਮ ਡਾਟਾ ਨੂੰ ਪ੍ਰਕ੍ਰਿਆ ਲਈ ਉਚਿਤ ਰੂਪ ਵਿੱਚ ਬਦਲਣਾ ਹੈ। ਸਾਡੇ ਮਾਮਲੇ ਵਿੱਚ, ਅਸੀਂ ਪੰਨਾ ਤੋਂ HTML ਸੋਰਸ ਕੋਡ ਡਾਊਨਲੋਡ ਕੀਤਾ ਹੈ, ਅਤੇ ਸਾਨੂੰ ਇਸ ਨੂੰ ਸਧਾਰਣ ਲਿਖਤ ਵਿੱਚ ਬਦਲਣਾ ਹੈ।

ਇਹ ਕਈ ਤਰੀਕਿਆਂ ਨਾਲ ਕੀਤਾ ਜਾ ਸਕਦਾ ਹੈ। ਅਸੀਂ [BeautifulSoup](https://www.crummy.com/software/BeautifulSoup/), ਜੋ ਕਿ HTML ਪਾਰਸਿੰਗ ਲਈ ਇੱਕ ਪ੍ਰਸਿੱਧ ਪਾਇਥਨ ਲਾਇਬ੍ਰੇਰੀ ਹੈ, ਵਰਤਾਂਗੇ। BeautifulSoup ਸਾਨੂੰ ਖਾਸ HTML ਤੱਤਾਂ 'ਤੇ ਧਿਆਨ ਕੇਂਦਰਿਤ ਕਰਨ ਦੀ ਆਗਿਆ ਦਿੰਦੀ ਹੈ, ਤਾਂ ਜੋ ਅਸੀਂ ਵਿੱਕੀਪੀਡੀਆ ਦੇ ਮੁੱਖ ਲੇਖ ਸਮੱਗਰੀ 'ਤੇ ਧਿਆਨ ਕੇਂਦਰਿਤ ਕਰ ਸਕੀਏ ਅਤੇ ਕੁਝ ਨੈਵਿਗੇਸ਼ਨ ਮੇਨੂ, ਸਾਈਡਬਾਰ, ਫੁੱਟਰ ਅਤੇ ਹੋਰ ਬੇਅਰਥ ਸਮੱਗਰੀ ਨੂੰ ਘਟਾ ਸਕੀਏ (ਅਜੇ ਵੀ ਕੁਝ ਸਧਾਰਣ ਲਿਖਤ ਬਾਕੀ ਰਹਿ ਸਕਦੀ ਹੈ)।


ਪਹਿਲਾਂ, ਸਾਨੂੰ HTML ਪਾਰਸਿੰਗ ਲਈ BeautifulSoup ਲਾਇਬ੍ਰੇਰੀ ਇੰਸਟਾਲ ਕਰਨੀ ਪਵੇਗੀ:


```python
import sys
!{sys.executable} -m pip install beautifulsoup4
```

```python
from bs4 import BeautifulSoup

# Parse the HTML content
soup = BeautifulSoup(text, 'html.parser')

# Extract only the main article content from Wikipedia
# Wikipedia uses 'mw-parser-output' class for the main article content
content = soup.find('div', class_='mw-parser-output')

def clean_wikipedia_content(content_node):
    """Remove common non-article elements from a Wikipedia content node."""
    # Strip jump links, navboxes, reference lists/superscripts, edit sections, TOC, sidebars, etc.
    selectors = [
        '.mw-jump-link',
        '.navbox',
        '.reflist',
        'sup.reference',
        '.mw-editsection',
        '.hatnote',
        '.metadata',
        '.infobox',
        '#toc',
        '.toc',
        '.sidebar',
    ]
    for selector in selectors:
        for el in content_node.select(selector):
            el.decompose()

if content:
    # Clean the content node to better approximate article text only.
    clean_wikipedia_content(content)
    text = content.get_text(separator=' ', strip=True)
    print(text[:1000])
else:
    print("Could not find main content. Using full page text.")
    text = soup.get_text(separator=' ', strip=True)
    print(text[:1000])
```

## ਕਦਮ 3: ਜਾਣਕਾਰੀਆਂ ਪ੍ਰਾਪਤ ਕਰਨਾ

ਸਭ ਤੋਂ ਮਹੱਤਵਪੂਰਣ ਕਦਮ ਹੈ ਸਾਡੇ ਡੇਟਾ ਨੂੰ ਇਸ ਤਰ੍ਹਾਂ ਤਬਦੀਲ ਕਰਨਾ ਜਿੱਥੋਂ ਅਸੀਂ ਜਾਣਕਾਰੀਆਂ ਪ੍ਰਾਪਤ ਕਰ ਸਕੀਏ। ਸਾਡੇ ਮਾਮਲੇ ਵਿੱਚ, ਅਸੀਂ ਲਿਖਤੋਂ ਕੁੰਜੀ ਸ਼ਬਦ ਕੱਢਣੇ ਚਾਹੁੰਦੇ ਹਾਂ, ਅਤੇ ਦੇਖਣਾ ਚਾਹੁੰਦੇ ਹਾਂ ਕਿ ਕਿਹੜੇ ਕੁੰਜੀ ਸ਼ਬਦ ਵਧੇਰੇ ਅਰਥਪੂਰਨ ਹਨ।

ਅਸੀਂ ਕੁੰਜੀ ਸ਼ਬਦ ਕੱਢਣ ਲਈ ਪਾਇਥਨ ਲਾਇਬ੍ਰੇਰੀ [RAKE](https://github.com/aneesha/RAKE) ਵਰਤਾਂਗੇ। ਸਭ ਤੋਂ ਪਹਿਲਾਂ, ਚਲੋ ਇਸ ਲਾਇਬ੍ਰੇਰੀ ਨੂੰ ਇੰਸਟਾਲ ਕਰ ਲਵਾਂ ਜੇਕਰ ਇਹ ਮੌਜੂਦ ਨਹੀਂ ਹੈ: 


```python
import sys
!{sys.executable} -m pip install nlp_rake
```

ਮੁੱਖ ਕਾਰਜਕੁਸ਼ਲਤਾ `Rake` ਆਬਜੈਕਟ ਤੋਂ ਉਪਲਬਧ ਹੈ, ਜਿਸ ਨੂੰ ਅਸੀਂ ਕੁਝ ਪੈਰਾਮੀਟਰਾਂ ਦੀ ਵਰਤੋਂ ਕਰਕੇ ਕਸਟਮਾਈਜ਼ ਕਰ ਸਕਦੇ ਹਾਂ। ਸਾਡੇ ਮਾਮਲੇ ਵਿੱਚ, ਅਸੀਂ ਕਿਸੇ ਕੀਵਰਡ ਦੀ ਘੱਟੋ-ਘੱਟ ਲੰਬਾਈ 5 ਅੱਖਰਾਂ ਤੇ ਸੈੱਟ ਕਰਾਂਗੇ, ਦਸਤਾਵੇਜ਼ ਵਿੱਚ ਕਿਸੇ ਕੀਵਰਡ ਦੀ ਘੱਟੋ-ਘੱਟ ਆਵ੍ਰਿਤੀ 3 ਤੇ ਅਤੇ ਕਿਸੇ ਕੀਵਰਡ ਵਿੱਚ ਵੱਧ ਤੋਂ ਵੱਧ ਸ਼ਬਦਾਂ ਦੀ ਗਿਣਤੀ 2 ਤੇ। ਹੋਰ ਮੁੱਲਾਂ ਨਾਲ ਖੇਡਣ ਲਈ ਸੁਤੰਤਰ ਰਹੋ ਅਤੇ ਨਤੀਜੇ ਦਾ ਅਵਲੋਕਨ ਕਰੋ।


```python
import nlp_rake
extractor = nlp_rake.Rake(max_words=2,min_freq=3,min_chars=5)
res = extractor.apply(text)
res
```


ਸਾਨੂੰ ਇੱਕ ਸੂਚੀ ਪ੍ਰਾਪਤ ਹੋਈ ਜਿਸ ਵਿੱਚ ਸ਼ਬਦਾਂ ਦੇ ਨਾਲ-ਨਾਲ ਉਨ੍ਹਾਂ ਦੀ ਮਹੱਤਤਾ ਦਾ ਦਰਜਾ ਵੀ ਦਿੱਤਾ ਗਿਆ ਹੈ। ਜਿਵੇਂ ਤੁਸੀਂ ਵੇਖ ਸਕਦੇ ਹੋ, ਸਭ ਤੋਂ ਸੰਬੰਧਿਤ ਵਿਸ਼ੇਸ਼ਤਾਵਾਂ, ਜਿਵੇਂ ਕਿ ਮਸ਼ੀਨ ਲਰਨਿੰਗ ਅਤੇ ਬਿਗ ਡਾਟਾ, ਸੂਚੀ ਵਿੱਚ ਸਿਖਰਲੇ ਸਥਾਨਾਂ 'ਤੇ ਮੌਜੂਦ ਹਨ।

## ਕਦਮ 4: ਨਤੀਜੇ ਦੀ ਵਿਜ਼ੂਅਲਾਈਜ਼ੇਸ਼ਨ

ਲੋਕ ਡਾਟਾ ਨੂੰ ਸਭ ਤੋਂ ਵਧੀਆ ਵਿਜ਼ੂਅਲ ਰੂਪ ਵਿੱਚ ਸਮਝ ਸਕਦੇ ਹਨ। ਇਸ ਲਈ ਡਾਟਾ ਨੂੰ ਵਿਜ਼ੂਅਲ ਬਣਾਉਣਾ ਬਹੁਤ ਵਾਰ ਜ਼ਰੂਰੀ ਹੁੰਦਾ ਹੈ ਤਾਂ ਜੋ ਕੁਝ ਅਹੰਕਾਰ ਪ੍ਰਾਪਤ ਕੀਤੇ ਜਾ ਸਕਣ। ਅਸੀਂ ਪਾਇਥਨ ਵਿੱਚ `matplotlib` ਲਾਇਬ੍ਰੇਰੀ ਦੀ ਵਰਤੋਂ ਕਰਕੇ ਕੀਵਰਡਸ ਦੀ ਸਧਾਰਣ ਵੰਡ ਨੂੰ ਉਨ੍ਹਾਂ ਦੀ ਸੰਬੰਧਿਤਤਾ ਦੇ ਨਾਲ ਪਲਾਟ ਕਰ ਸਕਦੇ ਹਾਂ:


```python
import matplotlib.pyplot as plt

def plot(pair_list):
    k,v = zip(*pair_list)
    plt.bar(range(len(k)),v)
    plt.xticks(range(len(k)),k,rotation='vertical')
    plt.show()

plot(res)
```

ਫਿਰ ਵੀ, ਸ਼ਬਦਾਂ ਦੀ ਆਵਿਰਤੀ ਦਿਖਾਉਣ ਦਾ ਇਕ ਹੋਰ ਵਧੀਆ ਤਰੀਕਾ ਹੈ - **ਵਰਡ ਕਲਾਉਡ** ਦੀ ਵਰਤੋਂ ਕਰਕੇ। ਸਾਨੂੰ apਣੀ ਕੀਵਰਡ ਸੂਚੀ ਵਿੱਚੋਂ ਵਰਡ ਕਲਾਉਡ ਬਣਾਉਣ ਲਈ ਇੱਕ ਹੋਰ ਲਾਇਬ੍ਰੇਰੀ ਇੰਸਟਾਲ ਕਰਨ ਦੀ ਲੋੜ ਪਵੇਗੀ।


```python
!{sys.executable} -m pip install wordcloud
```

`WordCloud` ਆਬਜੈਕਟ ਮੂਲ ਪਾਠ ਜਾਂ ਸ਼ਬਦਾਂ ਦੀ ਪਹਿਲਾਂ ਤੋਂ ਗਣਨਾ ਕੀਤੀ ਸੂਚੀ ਆਪਣੇ ਫ੍ਰੀਕਵੇਂਸੀਜ਼ ਦੇ ਨਾਲ ਲੈਣ ਅਤੇ ਇਕ ਚਿੱਤਰ ਵਾਪਸ ਕਰਨ ਲਈ ਜ਼ਿੰਮੇਵਾਰ ਹੈ, ਜਿਸ ਨੂੰ ਫਿਰ `matplotlib` ਦੀ ਵਰਤੋਂ ਨਾਲ ਦਰਸਾਇਆ ਜਾ ਸਕਦਾ ਹੈ:


```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

wc = WordCloud(background_color='white',width=800,height=600)
plt.figure(figsize=(15,7))
plt.imshow(wc.generate_from_frequencies({ k:v for k,v in res }))
```

ਅਸੀਂ ਮੂਲ ਪਾਠ ਨੂੰ ਵੀ `WordCloud` ਵਿੱਚ ਭੇਜ ਸਕਦੇ ਹਾਂ - ਚਲੋ ਵੇਖੀਏ ਕਿ ਕੀ ਅਸੀਂ ਸਮਾਨ ਨਤੀਜਾ ਪ੍ਰਾਪਤ ਕਰ ਸਕਦੇ ਹਾਂ:


```python
plt.figure(figsize=(15,7))
plt.imshow(wc.generate(text))
```

```python
wc.generate(text).to_file('images/ds_wordcloud.png')
```

ਤੁਸੀਂ ਦੇਖ ਸਕਦੇ ਹੋ ਕਿ ਵਰਡ ਕਲਾਉਡ ਹੁਣ ਜ਼ਿਆਦਾ ਪ੍ਰਭਾਵਸ਼ਾਲੀ ਲੱਗਦਾ ਹੈ, ਪਰ ਇਸ ਵਿੱਚ ਕਾਫੀ ਜ਼ਬਰਦਸਤ ਸ਼ੋਰ ਵੀ ਹੁੰਦਾ ਹੈ (ਜਿਵੇਂ ਕਿ `Retrieved on` ਵਰਗੇ ਬੇਸਮਝ ਸ਼ਬਦ). ਇਸਦੇ ਨਾਲ, ਸਾਨੂੰ ਥੋੜ੍ਹੇ ਕੁੰਜੀ ਸ਼ਬਦ ਮਿਲਦੇ ਹਨ ਜੋ ਦੋ ਸ਼ਬਦਾਂ ਦੇ ਬਣੇ ਹੁੰਦੇ ਹਨ, ਜਿਵੇਂ *ਡੇਟਾ ਸਾਇੰਟਿਸਟ* ਜਾਂ *ਕੰਪਿਊਟਰ ਸਾਇੰਸ*. ਇਹ ਇਸ ਲਈ ਹੈ ਕਿਉਂਕਿ RAKE ਅਲਗੋਰਿਦਮ ਲਿਖਤ ਵਿੱਚੋਂ ਚੰਗੇ ਕੁੰਜੀ ਸ਼ਬਦਾਂ ਦੀ ਚੋਣ ਕਰਨ ਵਿੱਚ ਕਾਫੀ ਵਧੀਆ ਕੰਮ ਕਰਦਾ ਹੈ। ਇਹ ਉਦਾਹਰਨ ਡੇਟਾ ਦੀ ਪਹਿਲਾ-ਪ੍ਰਕਿਆ ਅਤੇ ਸਾਫ-ਸੁਥਰਾ ਕਰਨ ਦੀ ਮਹੱਤਤਾ ਨੂੰ ਦਰਸਾਉਂਦੀ ਹੈ, ਕਿਉਂਕਿ ਆਖਿਰ ਵਿੱਚ ਸਾਫ-ਸੁਥਰਾ ਚਿੱਤਰ ਸਾਨੂੰ ਬਿਹਤਰ ਫੈਸਲੇ ਕਰਨ ਵਿੱਚ ਮਦਦ ਕਰੇਗਾ। 

ਇਸ ਅਭਿਆਸ ਵਿੱਚ ਅਸੀਂ ਵਿਕੀਪੀਡੀਆ ਲਿਖਤ ਵਿੱਚੋਂ ਕੁਝ ਅਰਥ ਕੱਢਨ ਦੀ ਸਧਾਰਨ ਪ੍ਰਕਿਰਿਆ ਵਿਚੋਂ ਗੁਜ਼ਰੇ ਹਾਂ, ਜਿਵੇਂ ਕਿ ਕੁੰਜੀ ਸ਼ਬਦ ਅਤੇ ਵਰਡ ਕਲਾਉਡ ਦੇ ਰੂਪ ਵਿੱਚ। ਇਹ ਉਦਾਹਰਨ ਕਾਫੀ ਸਧਾਰਨ ਹੈ, ਪਰ ਇਹ ਉਸ ਸਾਰੀ ਪ੍ਰਕਿਰਿਆ ਨੂੰ ਚੰਗੀ ਤਰ੍ਹਾਂ ਦਰਸਾਉਂਦਾ ਹੈ ਜੋ ਇਕ ਡੇਟਾ ਸਾਇੰਟਿਸਟ ਡੇਟਾ ਨਾਲ ਕੰਮ ਕਰਦਿਆਂ ਲੈਂਦਾ ਹੈ, ਡੇਟਾ ਪ੍ਰਾਪਤੀ ਤੋਂ ਸ਼ੁਰੂ ਕਰਕੇ ਵਿਜ਼ੁਅਲਾਈਜ਼ੇਸ਼ਨ ਤੱਕ। 

ਸਾਡੇ ਕੋਰਸ ਵਿੱਚ ਅਸੀਂ ਇਹ ਸਾਰੀਆਂ ਪ੍ਰਕਿਰਿਆਵਾਂ ਵਿਸਥਾਰ ਨਾਲ ਚਰਚਾ ਕਰਾਂਗੇ।
