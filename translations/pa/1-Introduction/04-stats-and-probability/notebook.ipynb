## ਭਰੋਸੇਯੋਗ ਦਾਇਰੇ

ਅਸੀਂ ਹੁਣ ਬੇਸਬਾਲ ਖਿਡਾਰੀਆਂ ਦੇ ਵਜ਼ਨ ਅਤੇ ਕੱਦ ਲਈ ਭਰੋਸੇਯੋਗ ਦਾਇਰੇ ਦੀ ਗਣਨਾ ਕਰਾਂਗੇ। ਅਸੀਂ ਕੋਡ ਦੀ ਵਰਤੋਂ ਕਰਾਂਗੇ [ਇਸ ਸਟੈਕਓਵਰਫਲੋ ਚਰਚਾ ਤੋਂ](https://stackoverflow.com/questions/15033511/compute-a-confidence-interval-from-sample-data):


```python
import scipy.stats

def mean_confidence_interval(data, confidence=0.95):
    a = 1.0 * np.array(data)
    n = len(a)
    m, se = np.mean(a), scipy.stats.sem(a)
    h = se * scipy.stats.t.ppf((1 + confidence) / 2., n-1)
    return m, h

for p in [0.85, 0.9, 0.95]:
    m, h = mean_confidence_interval(df['Weight'].ffill(),p)
    print(f"p={p:.2f}, mean = {m:.2f} ± {h:.2f}")
```

## ਹਿਪੋਥੈਸਿਸ ਟੈਸਟਿੰਗ

ਆਓ ਸਾਡੇ ਬੇਸਬॉल ਖਿਡਾਰੀਆਂ ਦੇ ਡੇਟਾਸੇਟ ਵਿੱਚ ਵੱਖ-ਵੱਖ ਭੂਮਿਕਾਵਾਂ ਦੀ ਖੋਜ ਕਰੀਏ:


```python
df.groupby('Role').agg({ 'Weight' : 'mean', 'Height' : 'mean', 'Age' : 'count'}).rename(columns={ 'Age' : 'Count'})
```

ਆਓ ਇਹ ਪਰਖੀਏ ਕਿ ਪਹਿਲੇ ਬੇਸਮੈਨ ਦੂਜੇ ਬੇਸਮੈਨ ਨਾਲੋਂ ਲੰਬੇ ਹੁੰਦੇ ਹਨ। ਇਸ ਦੀ ਸਾਬਤ ਕਰਨ ਦਾ ਸਭ ਤੋਂ ਸਾਦਾ ਤਰੀਕਾ ਇਹ ਹੈ ਕਿ ਅਸੀਂ ਕਾਨਫੀਡੈਂਸ ਇੰਟਰਵਲ ਨੂੰ ਟੈਸਟ ਕਰੀਏ:


```python
for p in [0.85,0.9,0.95]:
    m1, h1 = mean_confidence_interval(df.loc[df['Role']=='First_Baseman',['Height']],p)
    m2, h2 = mean_confidence_interval(df.loc[df['Role']=='Second_Baseman',['Height']],p)
    print(f'Conf={p:.2f}, 1st basemen height: {m1-h1[0]:.2f}..{m1+h1[0]:.2f}, 2nd basemen height: {m2-h2[0]:.2f}..{m2+h2[0]:.2f}')
```

ਅਸੀਂ ਦੇਖ ਸਕਦੇ ਹਾਂ ਕਿ ਅੰਤਰਾਲ ਇੱਕ ਦੂਜੇ ਨਾਲ ਨਹੀਂ ਮੀਲਦੇ।  

ਸਿਧਾਂਤ ਨੂੰ ਸਾਬਤ ਕਰਨ ਦਾ ਇੱਕ ਅੰਕੜਾ ਨੁਮਾ ਜ਼ਿਆਦਾ ਸਹੀ ਤਰੀਕਾ ਹੈ **ਸਟੂਡੈਂਟ t-ਟੈਸਟ** ਦੀ ਵਰਤੋਂ ਕਰਨੀ:  


```python
from scipy.stats import ttest_ind

tval, pval = ttest_ind(df.loc[df['Role']=='First_Baseman',['Height']], df.loc[df['Role']=='Second_Baseman',['Height']],equal_var=False)
print(f"T-value = {tval[0]:.2f}\nP-value: {pval[0]}")
```

`ttest_ind` ਫੰਕਸ਼ਨ ਵੱਲੋਂ ਵਾਪਸ ਕੀਤੀਆਂ ਦੋ ਮੁੱਲ ਹਨ:
* p-ਮੁੱਲ ਨੂੰ ਦੋ ਵੰਡਾਂ ਦੇ ਇੱਕੋ ਜਿਹੇ ਔਸਤ ਹੋਣ ਦੀ ਸੰਭਾਵਨਾ ਵਜੋਂ ਦੇਖਿਆ ਜਾ ਸਕਦਾ ਹੈ। ਸਾਡੇ ਮਾਮਲੇ ਵਿੱਚ, ਇਹ ਕਾਫੀ ਘੱਟ ਹੈ, ਜਿਸਦਾ ਅਰਥ ਹੈ ਕਿ ਪਹਿਲੇ ਬੇਸਮੈਨ ਲੰਬੇ ਹੋਣ ਦੇ ਮਜ਼ਬੂਤ ਸਬੂਤ ਹਨ।
* t-ਮੁੱਲ ਤੁਹਾਡੇ ਨਾਰਮਲਾਈਜ਼ਡ ਔਸਤ ਅੰਤਰ ਦਾ ਮੱਧਮ ਮੁੱਲ ਹੈ ਜੋ t-ਟੈਸਟ ਵਿੱਚ ਵਰਤਿਆ ਜਾਂਦਾ ਹੈ, ਅਤੇ ਇਸਨੂੰ ਦਿੱਤੀ ਗਈ ਭਰੋਸਾ ਕਦਰ ਲਈ ਇੱਕ ਸੀਮਾ ਮੁੱਲ ਨਾਲ ਤੁਲਨਾ ਕੀਤੀ ਜਾਂਦੀ ਹੈ।


## ਸੈਂਟ੍ਰਲ ਲਿਮਿਟ ਥਿਊਰਮ ਨਾਲ ਸਧਾਰਣ ਵੰਡ ਦੀ ਸਿਮੂਲੇਸ਼ਨ

ਪਾਇਥਨ ਵਿੱਚ ਛਦਮ-ਬੇਤਰਤੀਬ ਜੈਨਰੇਟਰ ਸਾਨੂੰ ਇੱਕ ਸਮਾਨ ਵੰਡ ਦੇਣ ਲਈ ਡਿਜ਼ਾਇਨ ਕੀਤਾ ਗਿਆ ਹੈ। ਜੇ ਅਸੀਂ ਸਧਾਰਣ ਵੰਡ ਲਈ ਜੈਨਰੇਟਰ ਬਣਾਉਣਾ ਚਾਹੁੰਦੇ ਹਾਂ, ਤਾਂ ਅਸੀਂ ਸੈਂਟ੍ਰਲ ਲਿਮਿਟ ਥਿਊਰਮ ਦੀ ਵਰਤੋਂ ਕਰ ਸਕਦੇ ਹਾਂ। ਇੱਕ ਸਧਾਰਣ ਵੰਡ ਵਾਲੀ ਕੀਮਤ ਪ੍ਰਾਪਤ ਕਰਨ ਲਈ ਅਸੀਂ ਸਿਰਫ਼ ਇਕ ਸਮਾਨ-ਜਨਰੇਟ ਕੀਤਾ ਗਿਆ ਨਮੂਨਾ ਦਾ ਮੱਧ (ਮੀਨ) ਕਾਲਕੁਲੇਟ ਕਰਾਂਗੇ। 


```python
def normal_random(sample_size=100):
    sample = [random.uniform(0,1) for _ in range(sample_size) ]
    return sum(sample)/sample_size

sample = [normal_random() for _ in range(100)]
plt.figure(figsize=(10,6))
plt.hist(sample)
plt.tight_layout()
plt.show()
```

## ਕੋਰਲੇਸ਼ਨ ਅਤੇ ਇਵਿਲ ਬੇਸਬਾਲ ਕਾਰਪੋਰੇਸ਼ਨ

ਕੋਰਲੇਸ਼ਨ ਸਾਨੂੰ ਡੇਟਾ ਕ੍ਰਮਾਂ ਵਿੱਚ ਸਬੰਧ ਲੱਭਣ ਦੀ ਆਗਿਆ ਦਿੰਦਾ ਹੈ। ਸਾਡੇ ਖਿਡੌਣੇ ਉਦਾਹਰਨ ਵਿੱਚ, ਆਓ ਧਰਾਈਏ ਕਿ ਇੱਕ ਬੁਰਾ ਬੇਸਬਾਲ ਕਾਰਪੋਰੇਸ਼ਨ ਹੈ ਜੋ ਆਪਣੇ ਖਿਡਾਰੀਆਂ ਨੂੰ ਉਨ੍ਹਾਂ ਦੀ ਲੰਬਾਈ ਅਨੁਸਾਰ ਤਨਖ਼ਾਹ ਦਿੰਦਾ ਹੈ - ਜਿੰਨਾ ਲੰਬਾ ਖਿਡਾਰੀ ਹੋਵੇ, ਉਸਨੂੰ ਉੰਨਾ ਜ਼ਿਆਦਾ ਪੈਸਾ ਮਿਲਦਾ ਹੈ। ਧਰੋ ਕਿ ਅੱਧਾਰ ਤਨਖ਼ਾਹ $1000 ਹੈ, ਅਤੇ ਲੰਬਾਈ ਅਨੁਸਾਰ $0 ਤੋਂ $100 ਤੱਕ ਅਤਿਰਕਤ ਬੋਨਸ ਮਿਲਦਾ ਹੈ। ਅਸੀਂ MLB ਦੇ ਅਸਲ ਖਿਡਾਰੀਆਂ ਨੂੰ ਲਵਾਂਗੇ ਅਤੇ ਉਨ੍ਹਾਂ ਦੀ ਕਲਪਨਾਤਮਕ ਤਨਖ਼ਾਹ ਦੀ ਗਿਣਤੀ ਕਰਾਂਗੇ:


```python
heights = df['Height'].ffill()
salaries = 1000+(heights-heights.min())/(heights.max()-heights.mean())*100
print(list(zip(heights, salaries))[:10])
```

ਆਓ ਹੁਣ ਉਹਨਾਂ ਕ੍ਰਮਾਂਕਾਂ ਦੀ ਕੋਵੈਰੀਅੰਸ ਅਤੇ ਕੋਰਿਲੇਸ਼ਨ ਦੀ ਗਣਨਾ ਕਰੀਏ। `np.cov` ਸਾਨੂੰ ਇਕ ਆਖਿਆ ਜਾ ਸਕਦਾ ਹੈ **ਕੋਵੈਰੀਅੰਸ ਮੈਟ੍ਰਿਕਸ**, ਜੋ ਕਿ ਕਈ ਚਰਾਂ ਲਈ ਕੋਵੈਰੀਅੰਸ ਦਾ ਵਿਸਥਾਰ ਹੈ। ਕੋਵੈਰੀਅੰਸ ਮੈਟ੍ਰਿਕਸ $M$ ਦਾ ਤੱਤ $M_{ij}$ ਇੰਪੁੱਟ ਚਰਾਂ $X_i$ ਅਤੇ $X_j$ ਵਿਚਕਾਰ ਕੋਰਿਲੇਸ਼ਨ ਹੈ, ਅਤੇ ਤੱਕਣੀ ਮੁੱਲ $M_{ii}$ $X_i$ ਦਾ ਵੇਰੀਐਂਸ ਹੈ। ਇਸੇ ਤਰ੍ਹਾਂ, `np.corrcoef` ਸਾਨੂੰ **ਕੋਰਿਲੇਸ਼ਨ ਮੈਟਰਿਕਸ** ਦੇਵੇਗਾ।


```python
print(f"Covariance matrix:\n{np.cov(heights, salaries)}")
print(f"Covariance = {np.cov(heights, salaries)[0,1]}")
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

1 ਦੇ ਬਰਾਬਰ ਸਹਿਸੰਬੰਧ ਦਾ ਅਰਥ ਹੈ ਕਿ ਦੋ ਚਲ-ਪేరਿਆਂ ਦਰਮਿਆਨ ਇਕ ਮਜ਼ਬੂਤ **ਰੇਖੀਅਤਮਕ ਸੰਬੰਧ** ਹੈ। ਅਸੀਂ ਇੱਕ ਮੁੱਲ ਨੂੰ ਦੂਜੇ ਦੇ ਖਿਲਾਬ ਪਲੌਟ ਕਰਕੇ ਰੇਖੀਅਤਮਕ ਸੰਬੰਧ ਨਜ਼ਰੀਅਤੀ ਤੌਰ 'ਤੇ ਦੇਖ ਸਕਦੇ ਹਾਂ:


```python
plt.figure(figsize=(10,6))
plt.scatter(heights,salaries)
plt.tight_layout()
plt.show()
```

ਆਓ ਵੇਖੀਏ ਕਿ ਕੀ ਹੁੰਦਾ ਹੈ ਜੇ ਸੰਬੰਧ ਰੇਖੀਅ ਨਹੀਂ ਹੈ। ਧਰੋ ਕਿ ਸਾਡੀ ਕੰਪਨੀ ਨੇ ਉਚਾਈਆਂ ਅਤੇ ਤਨਖਾਹਾਂ ਦੇ ਵਿਚਕਾਰ ਸਪੱਸ਼ਟ ਰੇਖੀਅ ਇੰਨਦਾ ਨੂੰ ਛੁਪਾ ਕੇ ਫਾਰਮੂਲਾ ਵਿੱਚ ਕੁਝ ਗੈਰ-ਰੇਖੀਅਤਾ ਜਿਵੇਂ ਕਿ `sin` ਸ਼ਾਮਲ ਕਰ ਦਿੱਤਾ: 


```python
salaries = 1000+np.sin((heights-heights.min())/(heights.max()-heights.mean()))*100
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

ਇਸ ਮਾਮਲੇ ਵਿੱਚ, ਸਮਬੰਧ ਥੋੜ੍ਹਾ ਛੋਟਾ ਹੈ, ਪਰ ਇਹ ਫਿਰ ਵੀ ਕਾਫੀ ਉੱਚਾ ਹੈ। ਹੁਣ, ਸੰਬੰਧ ਨੂੰ ਹੋਰ ਘੱਟ ਸਪੱਸ਼ਟ ਬਣਾਉਣ ਲਈ, ਅਸੀਂ ਸੇਲਰੀ ਵਿੱਚ ਕੁਝ ਅਜਿਹਾ ਯਾਦਰਸ਼ (randomness) ਸ਼ਾਮਿਲ ਕਰ ਸਕਦੇ ਹਾਂ ਜੋ ਕੁਝ ਬੇਤਹਾਸੀ ਪਰਿਵਰਤਨ ਵਾਲਾ ਹੋਵੇ। ਆਓ ਦੇਖੀਏ ਕਿ ਕੀ ਹੁੰਦਾ ਹੈ:


```python
salaries = 1000+np.sin((heights-heights.min())/(heights.max()-heights.mean()))*100+np.random.random(size=len(heights))*20-10
print(f"Correlation = {np.corrcoef(heights, salaries)[0,1]}")
```

```python
plt.figure(figsize=(10,6))
plt.scatter(heights, salaries)
plt.tight_layout()
plt.show()
```

> ਕੀ ਤੁਸੀਂ ਅੰਦਾਜ਼ਾ ਲਗਾ ਸਕਦੇ ਹੋ ਕਿ ਇਹ ਬਿੰਦੂ ਇਸ ਤਰ੍ਹਾਂ ਲੰਬਕਾਊ ਲਾਈਨਾਂ ਵਿੱਚ ਕਿਉਂ ਲੱਗਦੇ ਹਨ?

ਅਸੀਂ ਇੱਕ ਕ੍ਰਿਤ੍ਰਿਮ ਤਰੀਕੇ ਨਾਲ ਤਿਆਰ ਕੀਤੇ ਗਏ ਅਧਾਰ 'ਤਨਖਾਹ' ਅਤੇ ਪਰਖਿਆ ਗਿਆ ਵੈਰੀਏਬਲ *ਕੱਦ* ਦੇ ਵਿਚਕਾਰ ਸੰਬੰਧ ਦਾ ਅਧਿਐਨ ਕੀਤਾ ਹੈ। ਆਓ ਇਹ ਵੀ ਦੇਖੀਏ ਕਿ ਦੋ ਪੜ੍ਹੇ ਗਏ ਵੈਰੀਏਬਲ, ਜਿਵੇਂ ਕਿ ਕੱਦ ਅਤੇ ਵਜ਼ਨ, ਵੀ ਪਰਸਪਰ ਸੰਬੰਧਿਤ ਹਨ ਜਾਂ ਨਹੀਂ:


```python
np.corrcoef(df['Height'].ffill(),df['Weight'])
```

ਬਦਕਿਸਮਤੀ ਨਾਲ, ਸਾਨੂੰ ਕੋਈ ਨਤੀਜੇ ਨਹੀਂ ਮਿਲੇ - ਸਿਰਫ ਕੁਝ ਅਜੀਬ `nan` ਵੈਲਯੂਜ਼। ਇਹ ਇਸ ਕਰਕੇ ਹੋਇਆ ਹੈ ਕਿ ਸਾਡੇ ਸਿਰੀਜ਼ ਵਿੱਚ ਕੁਝ ਵੈਲਯੂਜ਼ ਅਪਰਿਭਾਸ਼ਿਤ ਹਨ, ਜੋ `nan` ਦੇ ਤੌਰ ਤੇ ਦਰਸਾਏ ਗਏ ਹਨ, ਜਿਸ ਕਾਰਨ ਆਪਰੇਸ਼ਨ ਦਾ ਨਤੀਜਾ ਵੀ ਅਪਰਿਭਾਸ਼ਿਤ ਹੋ ਜਾਂਦਾ ਹੈ। ਮੈਟ੍ਰਿਕਸ ਨੂੰ ਦੇਖ ਕੇ ਅਸੀਂ ਪਤਾ ਲਗਾ ਸਕਦੇ ਹਾਂ ਕਿ `Weight` ਸਮੱਸਿਆ ਵਾਲਾ ਕਾਲਮ ਹੈ, ਕਿਉਂਕਿ `Height` ਵੈਲਯੂਜ਼ ਵਿਚਕਾਰ ਸਵੈ-ਸੰਬੰਧ ਜੋੜਿਆ ਗਿਆ ਹੈ।

> ਇਹ ਉਦਾਹਰਨ **ਡਾਟਾ ਤਿਆਰੀ** ਅਤੇ **ਸਫਾਈ** ਦੀ ਮਹੱਤਤਾ ਦਿਖਾਉਂਦੀ ਹੈ। ਠੀਕ ਡਾਟਾ ਬਿਨਾਂ ਅਸੀਂ ਕੁਝ ਵੀ ਹਿਸਾਬ ਨਹੀਂ ਕਰ ਸਕਦੇ।

ਆਓ `fillna` ਮੈਥਡ ਦੀ ਵਰਤੋਂ ਕਰਕੇ ਗੁੰਮ ਵੈਲਯੂਜ਼ ਭਰੀਏ, ਅਤੇ ਸੰਬੰਧਤਾ ਦੀ ਗਣਨਾ ਕਰੀਏ: 


```python
np.corrcoef(df['Height'].ffill(), df['Weight'])
```

ਯਕੀਨਨ ਇਕ ਸੰਬੰਧ ਹੈ, ਪਰ ਸਾਡੇ ਕෘਤਰਿਮ ਉਦਾਹਰਨ ਵਾਂਗੁ ਇਨਾ ਮਜ਼ਬੂਤ ਨਹੀਂ। ਯਕੀਨਨ, ਜੇ ਅਸੀਂ ਇੱਕ ਮੁੱਲ ਨੂੰ ਦੂਜੇ ਦੇ ਖਿਲਾਫ ਸੈਂਟਰ ਪਲਾਟ 'ਤੇ ਦੇਖੀਏ, ਤਾਂ ਸੰਬੰਧ ਬਹੁਤ ਘੱਟ ਸਪਸ਼ਟ ਹੋਵੇਗਾ:


```python
plt.figure(figsize=(10,6))
plt.scatter(df['Weight'],df['Height'])
plt.xlabel('Weight')
plt.ylabel('Height')
plt.tight_layout()
plt.show()
```
